```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import rcParams
from matplotlib import rc
import matplotlib as mpl
import scipy.stats as stats
from scipy.integrate import quad
from pypopsyn.simulator.configuration import cfg
import pypopsyn.simulator.basics.constants as const
import utilities.plot_settings
from matplotlib.ticker import NullFormatter
```

```python
d_to_s = 24 * 3600
```

```python
data = pd.read_csv(
    "../examples/data/MagneticWhiteDwarfs_B_rot_mass2.csv",
    sep=",",   
)
data_mass = data[~data["Mass"].isin(["nan"])]
data.head()
```

```python
P = data["P_rot(d)"].to_numpy()
B = data["B(MG)"].to_numpy()
M = data_mass["Mass"].to_numpy().astype(np.float32)
M = M[~np.isnan(M)]

P = P * d_to_s
B = B * 1.e6

print(min(P), max(P))
cond = (P < 10**8) & (B < 2e9)

P = P[cond]
B = B[cond]

print(M)
```

```python
log_P_mean = np.mean(np.log10(P))
log_P_std = np.std(np.log10(P))
print(log_P_mean)
print(log_P_std)

log_B_mean = np.mean(np.log10(B))
log_B_std = np.std(np.log10(B))
print(log_B_mean)
print(log_B_std)

M_mean = np.mean(M)
M_std = np.std(M)

print(M_mean)
print(M_std)
```

```python
log_P_bins = np.linspace(1, 11, 11)
log_P_grid = np.linspace(1, 11, 100)

fig, ax = plt.subplots(figsize=(15,8))

ax.hist(
    np.log10(P),
    bins=log_P_bins,
    histtype="step",
    edgecolor="tab:blue",
    lw=4,
    ls="-",
)
ax.plot(
    log_P_grid, 
    60*stats.norm.pdf(log_P_grid, log_P_mean, log_P_std),
    color="tab:red",
    lw=4,
    ls="-",
)

plt.xlabel(r"$\log_{10} P$ [s]")
plt.ylabel(r"Number of WDs")
#plt.xlim(1e65,1e89)
#plt.ylim(1.e-9, 10)
#plt.xscale('log')
#plt.yscale('log')

plt.show()
```

```python
log_B_bins = np.linspace(3, 13, 11)
log_B_grid = np.linspace(3, 13, 100)

fig, ax = plt.subplots(figsize=(15,8))

ax.hist(
    np.log10(B),
    bins=log_B_bins,
    histtype="step",
    edgecolor="tab:blue",
    lw=4,
    ls="-",
)
ax.plot(
    log_B_grid, 
    60*stats.norm.pdf(log_B_grid, log_B_mean, log_B_std),
    color="tab:red",
    lw=4,
    ls="-",
)

plt.xlabel(r"$\log_{10} B$ [G]")
plt.ylabel(r"Number of WDs")
#plt.xlim(1e65,1e89)
#plt.ylim(1.e-9, 10)
#plt.xscale('log')
#plt.yscale('log')

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(12, 10))

ax.plot(
    P,
    B,
    linestyle="None",
    marker="o",
    color="tab:cyan",
    markersize=10,
    alpha=1,
    rasterized=True,
    label = 'initial population'
)

#plt.axvline(10**(-0.7))
ax.set_xlabel(r"$P$ [s]")
ax.set_ylabel(r"$B [G]$")
plt.xscale('log') 
plt.yscale('log') 

#plt.savefig()
plt.show()
```

```python
M_bins = np.linspace(0.5, 1.4, 11)
M_grid = np.linspace(0.5, 1.4, 100)

fig, ax = plt.subplots(figsize=(15,8))

ax.hist(
    M,
    bins=M_bins,
    histtype="step",
    edgecolor="tab:blue",
    lw=4,
    ls="-",
)
ax.plot(
    M_grid, 
    7*stats.norm.pdf(M_grid, M_mean, M_std),
    color="tab:red",
    lw=4,
    ls="-",
)

plt.xlabel(r"$Mass$ [M$_{\odot}$]")
plt.ylabel(r"Number of WDs")
#plt.xlim(1e65,1e89)
#plt.ylim(1.e-9, 10)
#plt.xscale('log')
#plt.yscale('log')

plt.show()
```

```python
log_P_bins = np.linspace(1, 8, 7)
log_B_bins = np.linspace(4, 10, 7)

nullfmt = NullFormatter()         # no labels

# definitions for the axes
left, width = 0.1, 0.65
bottom, height = 0.1, 0.65
bottom_h = left_h = left + width + 0.02

rect_scatter = [left, bottom, width, height]
rect_histx = [left, bottom_h, width, 0.2]
rect_histy = [left_h, bottom, 0.2, height]

# start with a rectangular Figure
fig = plt.figure(1, figsize=(12, 12))

ax_scatter = plt.axes(rect_scatter)
ax_histx = plt.axes(rect_histx)
ax_histy = plt.axes(rect_histy)

# no labels
ax_histx.xaxis.set_major_formatter(nullfmt)
ax_histy.yaxis.set_major_formatter(nullfmt)

# the scatter plot:
ax_scatter.scatter(np.log10(P), np.log10(B))

ax_scatter.set_xlim((1, 8))
ax_scatter.set_ylim((4, 10))

ax_histx.hist(np.log10(P), bins=log_P_bins)
ax_histx.plot(
    log_P_grid, 
    50*stats.norm.pdf(log_P_grid, log_P_mean, log_P_std),
    color="tab:red",
    lw=4,
    ls="-",
)

ax_histy.hist(np.log10(B), bins=log_B_bins, orientation='horizontal')
ax_histy.plot(
    50*stats.norm.pdf(log_B_grid, log_B_mean, log_B_std),
    log_B_grid, 
    color="tab:red",
    lw=4,
    ls="-",
)


ax_histx.set_xlim(1, 8)
ax_histy.set_ylim(4, 10)

# Set titles and labels
ax_scatter.set_xlabel(r'$\log_{10}{(P \, [s])}$')
ax_scatter.set_ylabel(r'$\log_{10}{(B \, [G])}$')
ax_histx.set_ylabel('# WDs')
ax_histy.set_xlabel('# WDs')

ax_histx.set_yticks([5, 10, 15])
ax_histy.set_xticks([5, 10, 15])

# Remove spacing between subplots
fig.subplots_adjust(wspace=0.01, hspace=0.01)

plt.show()
fig.savefig('MWD_B_P.pdf')
```
