# Inference results (1 parameter prediction)

Plot the inference results of a model architecture trained to predict a single parameter.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

import utilities.plot_settings
import utilities.statistics as stat
```

Read the file containing inference results for the $\sigma_{\rm k}$ parameter of the Maxwell distribution of kick velocities. Insert the path to the file.

```python
data = pd.read_csv("../tests/inference_test/1par_sigmak_inference_results_final.csv")
target_sigmak = data["target:sigma_k"].to_numpy()
prediction_sigmak = data["predicted:sigma_k"].to_numpy()
```

Compute the RMSE and MRE values for the total inferred datasets and as a function of the target values.

```python
# Compute the RMSE and MRE values on the total range of the target parameter.
RMSE_sigmak = np.sqrt(
    np.mean((prediction_sigmak - target_sigmak) ** 2)
)
MRE_sigmak = np.mean(
    abs((prediction_sigmak - target_sigmak) / target_sigmak)
)

# Compute the statistics as a function of the target values in equally spaced bins.
n_bins = 50
(
    bin_centers_sigmak,
    running_rmse_sigmak,
    running_average_sigmak,
    running_mre_sigmak,
) = stat.inference_running_stat(
    prediction_sigmak, target_sigmak, n_bins
)
```

Plot prediction vs target values.

```python
fig, ax = plt.subplots()

ax.set_xlabel(r"Target $\sigma_{\rm k}$ [km s$^{-1}$]")
ax.set_ylabel(r"Predicted $\sigma_{\rm k}$ [km s$^{-1}$]")

ax.scatter(
    target_sigmak,
    prediction_sigmak,
    linestyle="None",
    marker="o",
    facecolors="black",
    edgecolors="None",
    s=10,
    alpha=0.3,
    rasterized=True,
)
ax.plot(
    target_sigmak,
    target_sigmak,
    linestyle="-",
    linewidth=3,
    color="tab:red",
    alpha=1.0,
    rasterized=True,
    zorder=1,
)
plt.show()
```

Plot residuals along with the computed RMSE value over the whole parameter range (red dashed line), the RMSE as a function of the target value (orange region) and the average value of the residuals (blue line).

```python
fig, ax = plt.subplots()

ax.set_xlabel(r"$\sigma_{\rm k, T}$ [km s$^{-1}$]")
ax.set_ylabel(r"$\sigma_{\rm k, P} - \sigma_{\rm k, T}$ [km s$^{-1}$]")

ax.scatter(
    target_sigmak,
    prediction_sigmak - target_sigmak,
    linestyle="None",
    marker="o",
    facecolors="black",
    edgecolors="None",
    s=10,
    alpha=0.3,
    rasterized=True,
)
ax.plot(
    np.linspace(1.0, 700.0, n_bins),
    np.zeros(n_bins),
    linestyle="-",
    linewidth=3,
    color="tab:red",
    alpha=1.0,
    rasterized=True,
    zorder=1,
)
ax.plot(
    np.linspace(1.0, 700.0, n_bins),
    np.zeros(n_bins) - RMSE_sigmak,
    linestyle="--",
    linewidth=5,
    color="tab:red",
    alpha=1.0,
    rasterized=True,
    zorder=1,
)
ax.plot(
    np.linspace(1.0, 700.0, n_bins),
    np.zeros(n_bins) + RMSE_sigmak,
    linestyle="--",
    linewidth=5,
    color="tab:red",
    alpha=1.0,
    rasterized=True,
    zorder=1,
)
ax.plot(
    bin_centers_sigmak,
    running_average_sigmak,
    linestyle="-",
    linewidth=5,
    color="tab:blue",
    alpha=1.0,
    rasterized=True,
    zorder=1,
)
ax.fill_between(
    bin_centers_sigmak,
    np.zeros(len(bin_centers_sigmak)) - running_rmse_sigmak,
    np.zeros(len(bin_centers_sigmak)) + running_rmse_sigmak,
    facecolors="tab:orange",
    edgecolors="None",
    alpha=0.5,
    rasterized=True,
    zorder=0,
)

plt.show()
```

Plot absolute relative residuals along with the computed MRE value over the whole parameter range (red dashed line), the MRE as a function of the target value (orange region) and the average value of the residuals (blue line).

```python
fig, ax = plt.subplots()

ax.set_yscale("log")
ax.set_xlabel(r"$\sigma_{\rm k, T}$ [km s$^{-1}$]")
ax.set_ylabel(
    r"$\left| (\sigma_{\rm k, P} - \sigma_{\rm k, T}) / \sigma_{\rm k, T} \right|$"
)

ax.scatter(
    target_sigmak,
    abs((prediction_sigmak - target_sigmak) / target_sigmak),
    linestyle="None",
    marker="o",
    facecolors="black",
    edgecolors="None",
    s=10,
    alpha=0.3,
    rasterized=True,
)
ax.plot(
    np.linspace(1.0, 700.0, 50),
    np.zeros(50) + MRE_sigmak,
    linestyle="--",
    linewidth=5,
    color="tab:red",
    alpha=1.0,
    rasterized=True,
    zorder=1,
)
ax.plot(
    bin_centers_sigmak,
    np.zeros(len(bin_centers_sigmak)) + running_mre_sigmak,
    linestyle="-",
    linewidth=5,
    color="tab:blue",
    alpha=1.0,
    rasterized=True,
    zorder=1,
)
plt.show()
```
