## DEATHLINE

```python
def first_dl_chen(P,R):
    R = np.ones(len(P))*R
    return (1/4)*(106.37 - 9.5*np.log10(R) + 7.5*np.log10(P))
def second_dl_chen(P,R):
    R = np.ones(len(P))*R
    return (1/4)*(96.692 - 8.5*np.log10(R) + 6.5*np.log10(P))
def third_dl_chen(P,R):
    R = np.ones(len(P))*R
    return (1/7)*(180.078 - 17*np.log10(R) + 13*np.log10(P))
def fourth_dl_chen(P,R):
    R = np.ones(len(P))*R
    return (1/4)*(91.85 - 8*np.log10(R) + 6*np.log10(P))
def first_dl_zhang(P,R):
    R = np.ones(len(P))*R
    return 25.96 - (9/4)*np.log10(R) + (7/4)*np.log10(P)
def second_dl_zhang(P,R):
    R = np.ones(len(P))*R
    return 23.54 - 2*np.log10(R) + (3/2)*np.log10(P)
```
