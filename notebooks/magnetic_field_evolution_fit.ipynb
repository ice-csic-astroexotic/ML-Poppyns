# Fit an analytical model to the simulated magnetic field evolution

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl

from pypopsyn.simulator.configuration import cfg
import pypopsyn.simulator.magneto_rotational_physics.magneto_rotational_evolution_fit as mre
import utilities.plot_settings
```

## Load the simulated magnetic field evolution curves

For the magneto-thermal simulation the following setup has been employed.
The equation of state is Sly4 EoS, the NS mass is $1.4 M_{\odot}$, the radius is $11.74$ km and the impurity
in the pasta layer is fixed to 100. Instead, for the impurity in the outer crust and the inner crust
(excluding pasta layer), the fit of [Carreau et al.(2020)](https://ui.adsabs.harvard.edu/abs/2020A%26A...640A..77C/abstract) has been used (see figure 5 in that paper).
The envelope model is taken from [Potekhin et al. (2015)](https://ui.adsabs.harvard.edu/abs/2015SSRv..191..239P/abstract).
The gaps are taken from [Andersson et al. (2005)](https://ui.adsabs.harvard.edu/abs/2015SSRv..191..239P/abstract) and are the following:
- SFB, Superfluid model for n crust [0=deactivate] (see superfluid.f in the magneto-thermal code)
- TToa, Superfluid model for n core [0=deactivate] (see superfluid.f in the magneto-thermal code)
- CCDKp, Superfluid model for p core [0=deactivate] (see superfluid.f in the magneto-thermal code)

The initial magnetic field of the simulated curves are: $10^{12}$ G, $10^{13}$ G, $10^{14}$ G, $10^{15}$ G, $3 \times 10^{15}$ G,

```python
t12_sim = df_simB12["t[yr]"].to_numpy().astype(np.float64) 
t13_sim = df_simB13["t[yr]"].to_numpy().astype(np.float64) 
t14_sim = df_simB14["t[yr]"].to_numpy().astype(np.float64) 
t15_sim = df_simB15["t[yr]"].to_numpy().astype(np.float64) 
t3e15_sim = df_simB3e15["t[yr]"].to_numpy().astype(np.float64) 

B12_sim = df_simB12["B[G]"].to_numpy().astype(np.float64) 
B13_sim = df_simB13["B[G]"].to_numpy().astype(np.float64) 
B14_sim = df_simB14["B[G]"].to_numpy().astype(np.float64) 
B15_sim = df_simB15["B[G]"].to_numpy().astype(np.float64) 
B3e15_sim = df_simB3e15["B[G]"].to_numpy().astype(np.float64)
```

## Evolution curves from the analytical fit model

```python
log_B_initial = np.array([10, 11, 12, 13, 14, 15, np.log10(3.e15), 16, 17])
B_initial = 10**log_B_initial
time = np.logspace(0., 9., 100)

B_asymptotic = 10 ** np.random.normal(
        cfg["B_millisec_mean"], cfg["B_millisec_sigma"], len(B_initial)
    )

B10_fit = mre.magnetic_field_evolution_fit_numpy(B_initial[0], time, B_asymptotic[0])
B11_fit = mre.magnetic_field_evolution_fit_numpy(B_initial[1], time, B_asymptotic[1])
B12_fit = mre.magnetic_field_evolution_fit_numpy(B_initial[2], time, B_asymptotic[2])
B13_fit = mre.magnetic_field_evolution_fit_numpy(B_initial[3], time, B_asymptotic[3])
B14_fit = mre.magnetic_field_evolution_fit_numpy(B_initial[4], time, B_asymptotic[4])
B15_fit = mre.magnetic_field_evolution_fit_numpy(B_initial[5], time, B_asymptotic[5])
B3e15_fit = mre.magnetic_field_evolution_fit_numpy(B_initial[6], time, B_asymptotic[6])
B16_fit = mre.magnetic_field_evolution_fit_numpy(B_initial[7], time, B_asymptotic[7])
B17_fit = mre.magnetic_field_evolution_fit_numpy(B_initial[8], time, B_asymptotic[8])
```

```python
# plot the magnetic field evolution curves
fig, ax = plt.subplots(figsize=(12,10))

colors = plt.get_cmap("viridis", len(log_B_initial))
norm = mpl.colors.Normalize(vmin=np.min(log_B_initial), vmax=np.max(log_B_initial)+1)
sm = plt.cm.ScalarMappable(norm=norm, cmap=colors)
sm.set_array([])

ax.set_xscale('log') 
ax.set_yscale('log')
#ax.set_xlim(1.,1.e5) 
#ax.set_ylim(1.e7,1.e17) 
ax.set_xlabel(r'Time [yr]')
ax.set_ylabel(r'$B_{\rm dip}$ [G]')

ax.plot( 
    t12_sim,
    B12_sim,
    linestyle='-',
    linewidth=4,
    color=colors(2),
    rasterized=True,
)
ax.plot( 
    t13_sim,
    B13_sim,
    linestyle='-',
    linewidth=4,
    color=colors(3),
    rasterized=True,
)
ax.plot( 
    t14_sim,
    B14_sim,
    linestyle='-',
    linewidth=4,
    color=colors(4),
    rasterized=True,
)
ax.plot( 
    t15_sim,
    B15_sim,
    linestyle='-',
    linewidth=4,
    color=colors(5),
    rasterized=True,
)
ax.plot( 
    t3e15_sim,
    B3e15_sim,
    linestyle='-',
    linewidth=4,
    color=colors(6),
    rasterized=True,
)

ax.plot( 
    time,
    B10_fit,
    linestyle='--',
    linewidth=4,
    color=colors(0),
    rasterized=True,
)

ax.plot( 
    time,
    B11_fit,
    linestyle='--',
    linewidth=4,
    color=colors(1),
    rasterized=True,
)
ax.plot( 
    time,
    B12_fit,
    linestyle='--',
    linewidth=4,
    color=colors(2),
    rasterized=True,
)
ax.plot( 
    time,
    B13_fit,
    linestyle='--',
    linewidth=4,
    color=colors(3),
    rasterized=True,
)
ax.plot( 
    time,
    B14_fit,
    linestyle='--',
    linewidth=4,
    color=colors(4),
    rasterized=True,
)
ax.plot( 
    time,
    B15_fit,
    linestyle='--',
    linewidth=4,
    color=colors(5),
    rasterized=True,
)
ax.plot( 
    time,
    B3e15_fit,
    linestyle='--',
    linewidth=4,
    color=colors(6),
    rasterized=True,
)
ax.plot( 
    time,
    B16_fit,
    linestyle='--',
    linewidth=4,
    color=colors(7),
    rasterized=True,
)
ax.plot( 
    time,
    B17_fit,
    linestyle='--',
    linewidth=4,
    color=colors(8),
    rasterized=True,
)



plt.show()
```
