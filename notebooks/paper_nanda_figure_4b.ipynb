```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import rcParams
from matplotlib import rc
import matplotlib as mpl
import scipy.stats as stats
from scipy.integrate import quad
from pypopsyn.simulator.configuration import cfg
import pypopsyn.simulator.basics.constants as const
import utilities.plot_settings
from matplotlib.collections import PathCollection
from matplotlib.legend_handler import HandlerPathCollection, HandlerLine2D

def update(handle, orig):
    handle.update_from(orig)
    handle.set_alpha(1)
```

```python
data_wd1 = pd.read_pickle(
    "../toremove/nanda_experiment/WD_exp3/final_population.pkl.gz",
    compression="gzip",
)
data_wd1.head()
```

```python
P_wd1 = data_wd1["P"]["[s]"].to_numpy()
P_dot_wd1 = data_wd1["P_dot"]["[s s^-1]"].to_numpy()
L_radio_bol_wd1 = data_wd1["L_radio_bol"]["[erg s^-1]"].to_numpy()
B_wd1 = data_wd1["B"]["[G]"].to_numpy()

intercept_radio_wd1 = data_wd1[('intercepted_radio',' ')] == 1
```

```python
data_wd2 = pd.read_pickle(
    "../toremove/nanda_experiment/WD_exp4/final_population.pkl.gz",
    compression="gzip",
)
data_wd2.head()
```

```python
P_wd2 = data_wd2["P"]["[s]"].to_numpy()
P_dot_wd2 = data_wd2["P_dot"]["[s s^-1]"].to_numpy()
L_radio_bol_wd2 = data_wd2["L_radio_bol"]["[erg s^-1]"].to_numpy()
B_wd1 = data_wd2["B"]["[G]"].to_numpy()

intercept_radio_wd2 = data_wd2[('intercepted_radio',' ')] == 1
```

```python
P_bins_wd = np.logspace(-2, 7, 31)

fig, ax = plt.subplots(figsize=(15,8))
#hist_P_i, bin_edges_P_i = np.histogram(P_i, bins = P_bins)

ax.hist(
    P_wd1,
    bins=P_bins_wd,
    histtype="step",
    edgecolor="lightskyblue",
    lw=4,
    ls="--",
    alpha=1,
    label="Final population WD3",
    cumulative=True,
)

ax.hist(
    P_wd1[intercept_radio_wd1],
    bins=P_bins_wd,
    histtype="step",
    edgecolor="tab:blue",
    lw=4,
    ls="-",
    alpha=1,
    label="Intercepting our LOS WD3",
    cumulative=True,
)
ax.hist(
    P_wd2,
    bins=P_bins_wd,
    histtype="step",
    edgecolor="peru",
    lw=4,
    ls="--",
    alpha=1,
    label="Final population WD4",
    cumulative=True,
)

ax.hist(
    P_wd2[intercept_radio_wd2],
    bins=P_bins_wd,
    histtype="step",
    edgecolor="tab:red",
    lw=4,
    ls="-",
    alpha=1,
    label="Intercepting our LOS WD4",
    cumulative=True,
)


plt.xlabel(r"$P$ [s]")
plt.ylabel(r"Number of WDs $(> P)$")
plt.xlim(1e-2,1e7)
#plt.ylim(1.e-9, 10)
plt.xscale('log')
plt.yscale('log')
plt.legend(frameon=False, loc=2, fontsize=20)

plt.grid()
plt.show()
```
