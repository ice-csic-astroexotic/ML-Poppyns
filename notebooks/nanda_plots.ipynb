```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import rcParams
from matplotlib import rc
import matplotlib as mpl
from scipy.integrate import quad
from pypopsyn.simulator.configuration import cfg
import pypopsyn.simulator.basics.constants as const
import utilities.plot_settings
from matplotlib.collections import PathCollection
from matplotlib.legend_handler import HandlerPathCollection, HandlerLine2D

def update(handle, orig):
    handle.update_from(orig)
    handle.set_alpha(1)
```

## no power-law component

```python
data_i_no_power = pd.read_pickle(
    "../toremove/nanda_experiment/no_powerlaw/initial_population.pkl.gz",
    compression="gzip",
)

data_f_no_power = pd.read_pickle(
    "../toremove/nanda_experiment/no_powerlaw/final_population.pkl.gz",
    compression="gzip",
)
data_f_no_power.head()
```

```python
P_i_no_power = data_i_no_power["P"]["[s]"].to_numpy()
P_dot_i_no_power = data_i_no_power["P_dot"]["[s s^-1]"].to_numpy()

P_f_no_power = data_f_no_power["P"]["[s]"].to_numpy()
P_dot_f_no_power = data_f_no_power["P_dot"]["[s s^-1]"].to_numpy()
L_radio_bol_no_power = data_f_no_power["L_radio_bol"]["[erg s^-1]"].to_numpy()
d_no_power = data_f_no_power["d"]["[kpc]"].to_numpy()
```

```python
fig, ax = plt.subplots(figsize=(12, 10))

ax.plot(
    P_i_no_power,
    P_dot_i_no_power,
    linestyle="None",
    marker="o",
    color="tab:cyan",
    markersize=1,
    alpha=1,
    rasterized=True,
    label = 'initial population'
)

ax.plot(
    P_f_no_power,
    P_dot_f_no_power,
    linestyle="None",
    marker="o",
    color="tab:blue",
    markersize=1,
    alpha=1,
    rasterized=True,
    label = 'final population'

)

#plt.axvline(10**(-0.7))
ax.set_xlabel(r"$P$ [s]")
ax.set_ylabel(r"$\dot{P}$")
plt.xscale('log') 
plt.yscale('log') 
plt.xlim(1e-2,2e4)
plt.ylim(1e-23,1e-5)
#plt.legend(bbox_to_anchor=(1, 1), frameon=False, loc=0, fontsize=20,markerscale=5)
plt.legend(frameon=False, loc=0, fontsize=20,markerscale=5, handler_map={PathCollection : HandlerPathCollection(update_func= update),
                        plt.Line2D : HandlerLine2D(update_func = update)})

#plt.savefig()
plt.show()
```

```python
P_bins_no_power = np.logspace(-3., 5., 31)

#P_bins_shifted = np.roll(P_bins, 1)
#size_bin = P_bins - P_bins_shifted

fig, ax = plt.subplots(figsize=(15,8))
#hist_P_i, bin_edges_P_i = np.histogram(P_i, bins = P_bins)

ax.hist(
    P_i_no_power,
    bins=P_bins_no_power,
    histtype="step",
    edgecolor="tab:cyan",
    lw=4,
    ls=":",
    label="Initial all",
)
ax.hist(
    P_f_no_power,
    bins=P_bins_no_power,
    histtype="step",
    edgecolor="tab:blue",
    lw=4,
    ls="-",
    label="Final all",
)

plt.xlabel(r"$P$ [s]")
plt.ylabel(r"Number of NSs")
plt.xlim(1e-2,2e4)
#plt.ylim(1.e-9, 10)
plt.xscale('log')
plt.yscale('log')
plt.legend(frameon=False, loc=0, fontsize=20)

plt.show()
```

```python
np.count_nonzero(P_f_no_power>1e3)
```

```python
fig, ax = plt.subplots(figsize=(12, 10))

ax.plot(
    P_f_no_power,
    L_radio_bol_no_power,
    linestyle="None",
    marker="o",
    color="tab:blue",
    markersize=1,
    alpha=1,
    rasterized=True
)

ax.set_xlabel(r"$P$ [s]")
ax.set_ylabel(r"$L_{\rm radio}$")
plt.xscale('log') 
plt.yscale('log') 
plt.xlim(1e-2,2e4)
plt.ylim(1e19,1e33)

plt.show()
```

## power-law -1

```python
data_i_power_1 = pd.read_pickle(
    "../toremove/nanda_experiment/powerlaw-1/initial_population.pkl.gz",
    compression="gzip",
)

data_f_power_1 = pd.read_pickle(
    "../toremove/nanda_experiment/powerlaw-1/final_population.pkl.gz",
    compression="gzip",
)
data_f_power_1.head()
```

```python
P_i_power_1 = data_i_power_1["P"]["[s]"].to_numpy()
P_dot_i_power_1 = data_i_power_1["P_dot"]["[s s^-1]"].to_numpy()

P_f_power_1 = data_f_power_1["P"]["[s]"].to_numpy()
P_dot_f_power_1 = data_f_power_1["P_dot"]["[s s^-1]"].to_numpy()
L_radio_bol_power_1 = data_f_power_1["L_radio_bol"]["[erg s^-1]"].to_numpy()
d_power_1 = data_f_power_1["d"]["[kpc]"].to_numpy()
```

```python
mask_log_normal_power_1 = (data_f_power_1['lognormal_flag'] == 1) 
mask_powerlaw_power_1 = (data_f_power_1['lognormal_flag'] == 0) 
```

```python
P_bins_power_1 = np.logspace(-3., 5., 31)

fig, ax = plt.subplots(figsize=(15,8))
#hist_P_i_power_1, bin_edges_P_i = np.histogram(P_i, bins = P_bins)

ax.hist(
    P_i_power_1[np.transpose(mask_log_normal_power_1.to_numpy())[0]],
    bins=P_bins_power_1,
    histtype="step",
    color="tab:cyan",
    lw=4,
    ls=":",
    label='log-normal initial period',
    density=True,
)
ax.hist(
    P_i_power_1[np.transpose(mask_powerlaw_power_1.to_numpy())[0]],
    bins=P_bins_power_1,
    histtype="step",
    edgecolor="tab:orange",
    lw=4,
    ls=":",
    label='powerlaw initial period',
    density=True,
)
ax.hist(
    P_i_power_1,
    bins=P_bins_power_1,
    histtype="step",
    edgecolor="tab:pink",
    lw=6,
    label='combined initial period distribution',
    density=True,
)

plt.xlabel(r"$P$ [s]")
plt.ylabel(r"Number of NSs")
plt.xscale('log')
plt.yscale('log')
plt.legend(frameon=False, loc=0, fontsize=20)
plt.xlim(1e-2,2e4)
plt.ylim(1.e-9, 10)

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(12, 10))

ax.plot(
    P_i_power_1[np.transpose(mask_log_normal_power_1.to_numpy())[0]],
    P_dot_i_power_1[np.transpose(mask_log_normal_power_1.to_numpy())[0]],
    linestyle="None",
    marker="o",
    color="tab:cyan",
    markersize=1,
    alpha=0.2,
    rasterized=True,
    label = 'log-normal initial period'
)

ax.plot(
    P_i_power_1[np.transpose(mask_powerlaw_power_1.to_numpy())[0]],
    P_dot_i_power_1[np.transpose(mask_powerlaw_power_1.to_numpy())[0]],
    linestyle="None",
    marker="o",
    color="tab:orange",
    markersize=1,
    alpha=0.2,
    rasterized=True,
    label = 'powerlaw initial period'
)


ax.plot(
    P_f_power_1[np.transpose(mask_log_normal_power_1.to_numpy())[0]],
    P_dot_f_power_1[np.transpose(mask_log_normal_power_1.to_numpy())[0]],
    linestyle="None",
    marker="o",
    color="tab:blue",
    markersize=1,
    alpha=0.2,
    rasterized=True,
    label = 'log-normal final period'

)

ax.plot(
    P_f_power_1[np.transpose(mask_powerlaw_power_1.to_numpy())[0]],
    P_dot_f_power_1[np.transpose(mask_powerlaw_power_1.to_numpy())[0]],
    linestyle="None",
    marker="o",
    color="tab:red",
    markersize=1,
    alpha=0.2,
    rasterized=True,
    label = 'power-law final period'

)
#plt.axvline(10**(-0.7))
ax.set_xlabel(r"$P$ [s]")
ax.set_ylabel(r"$\dot{P}$")
plt.xscale('log') 
plt.yscale('log') 
plt.ylim(1e-23,1e-5)
plt.xlim(1e-2,2e4)
plt.legend(loc=0, fontsize=20,markerscale=5, handler_map={PathCollection : HandlerPathCollection(update_func= update),
                        plt.Line2D : HandlerLine2D(update_func = update)})

#plt.savefig()
plt.show()
```

```python
P_i_w_int_01 = data_i_w_int_01["P"]["[s]"].to_numpy()
P_dot_i_w_int_01 = data_i_w_int_01["P_dot"]["[s s^-1]"].to_numpy()

P_f_w_int_01 = data_f_w_int_01["P"]["[s]"].to_numpy()
P_dot_f_w_int_01 = data_f_w_int_01["P_dot"]["[s s^-1]"].to_numpy()
L_radio_bol_w_int_01 = data_f_w_int_01["L_radio_bol"]["[erg s^-1]"].to_numpy()
d_w_int_01 = data_f_w_int_01["d"]["[kpc]"].to_numpy()

mask_log_normal_w_int_01 = (data_f_w_int_01['lognormal_flag'] == 1) 
mask_powerlaw_w_int_01 = (data_f_w_int_01['lognormal_flag'] == 0) 
```

```python
intercept_radio_w_int_01 = data_f_w_int_01[('intercepted_radio',' ')] == 1

fig, ax = plt.subplots(figsize=(12, 10))


ax.plot(
    P_f_w_int_01,
    P_dot_f_w_int_01,
    linestyle="None",
    marker="o",
    color="tab:blue",
    markersize=1,
    alpha=0.2,
    rasterized=True,
    label = 'final populations'

)

ax.plot(
    P_f_w_int_01[intercept_radio_w_int_01],
    P_dot_f_w_int_01[intercept_radio_w_int_01],
    linestyle="None",
    marker="o",
    color="tab:green",
    markersize=1,
    alpha=0.5,
    rasterized=True,
    label = 'intercept our los'

)
#plt.axvline(10**(-0.7))
ax.set_xlabel(r"$P$ [s]")
ax.set_ylabel(r"$\dot{P}$")
plt.xscale('log') 
plt.yscale('log') 
plt.ylim(1e-23,1e-5)
plt.xlim(1e-2,2e4)
plt.legend(loc=0, fontsize=20,markerscale=5, handler_map={PathCollection : HandlerPathCollection(update_func= update),
                        plt.Line2D : HandlerLine2D(update_func = update)})

#plt.savefig()
plt.show()
```

```python
P_bins_w_int_01 = np.logspace(-3., 5., 31)

fig, ax = plt.subplots(figsize=(15,8))
#hist_P_i, bin_edges_P_i = np.histogram(P_i, bins = P_bins)

ax.hist(
    P_f_w_int_01,
    bins=P_bins_w_int_01,
    histtype="step",
    edgecolor="tab:blue",
    lw=4,
    ls="-",
    label='final population',
    #density=True,
)
ax.hist(
    P_f_w_int_01[intercept_radio_w_int_01],
    bins=P_bins_w_int_01,
    histtype="step",
    edgecolor="tab:green",
    lw=4,
    ls="-",
    label='crossing our LOS',
    #density=True,
)

plt.xlabel(r"$P$ [s]")
plt.ylabel(r"Number of NSs")
plt.xlim(1e-2,2e4)
#plt.ylim(1.e-9, 10)
plt.xscale('log')
plt.yscale('log')
plt.legend(frameon=False, loc=0, fontsize=20)

plt.show()
```
