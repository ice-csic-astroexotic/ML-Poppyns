```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import rcParams
from matplotlib import rc
import matplotlib as mpl
from scipy.integrate import quad
from pypopsyn.simulator.configuration import cfg
import pypopsyn.simulator.basics.constants as const
import utilities.plot_settings
from matplotlib.collections import PathCollection
from matplotlib.legend_handler import HandlerPathCollection, HandlerLine2D

def update(handle, orig):
    handle.update_from(orig)
    handle.set_alpha(1)
```

## no power-law component

```python
data_i = pd.read_pickle(
    "../toremove/nanda_experiment/no_powerlaw/initial_population.pkl.gz",
    compression="gzip",
)

data_f = pd.read_pickle(
    "../toremove/nanda_experiment/no_powerlaw/final_population.pkl.gz",
    compression="gzip",
)
data_f.head()
```

```python
P_i = data_i["P"]["[s]"].to_numpy()
P_dot_i = data_i["P_dot"]["[s s^-1]"].to_numpy()

P_f = data_f["P"]["[s]"].to_numpy()
P_dot_f = data_f["P_dot"]["[s s^-1]"].to_numpy()
L_radio_bol = data_f["L_radio_bol"]["[erg s^-1]"].to_numpy()
d = data_f["d"]["[kpc]"].to_numpy()
```

```python
fig, ax = plt.subplots(figsize=(12, 10))

ax.plot(
    P_i,
    P_dot_i,
    linestyle="None",
    marker="o",
    color="tab:cyan",
    markersize=1,
    alpha=0.2,
    rasterized=True,
    label = 'initial population'
)

ax.plot(
    P_f,
    P_dot_f,
    linestyle="None",
    marker="o",
    color="tab:orange",
    markersize=1,
    alpha=0.2,
    rasterized=True,
    label = 'final population'

)

#plt.axvline(10**(-0.7))
ax.set_xlabel(r"$P$ [s]")
ax.set_ylabel(r"$\dot{P}$")
plt.xscale('log') 
plt.yscale('log') 
#plt.legend(bbox_to_anchor=(1, 1), frameon=False, loc=0, fontsize=20,markerscale=5)
plt.legend(bbox_to_anchor=(1, 1), frameon=False, loc=0, fontsize=20,markerscale=5, handler_map={PathCollection : HandlerPathCollection(update_func= update),
                        plt.Line2D : HandlerLine2D(update_func = update)})

#plt.savefig()
plt.show()
```

```python
P_bins = np.logspace(-3., 5., 31)

#P_bins_shifted = np.roll(P_bins, 1)
#size_bin = P_bins - P_bins_shifted

fig, ax = plt.subplots(figsize=(15,8))
#hist_P_i, bin_edges_P_i = np.histogram(P_i, bins = P_bins)

ax.hist(
    P_i,
    bins=P_bins,
    histtype="step",
    edgecolor="tab:blue",
    lw=4,
    ls="-",
    label="Initial all",
)
ax.hist(
    P_f,
    bins=P_bins,
    histtype="step",
    edgecolor="tab:red",
    lw=4,
    ls="-",
    label="Final all",
)

plt.xlabel(r"$P$ [s]")
plt.ylabel(r"Number of NSs")
#plt.xlim(0., 50.0)
#plt.ylim(0.1, 2.e5)
plt.xscale('log')
plt.yscale('log')
plt.legend(bbox_to_anchor=(1.05, 1), frameon=False, loc=0, fontsize=20)

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(12, 10))

ax.plot(
    P_f,
    L_radio_bol,
    linestyle="None",
    marker="o",
    color="tab:cyan",
    markersize=1,
    alpha=0.1,
    rasterized=True
)

ax.set_xlabel(r"$P$ [s]")
ax.set_ylabel(r"$L_{\rm radio}$")
plt.xscale('log') 
plt.yscale('log') 

plt.show()
```

## power-law -1

```python
data_i = pd.read_pickle(
    "../toremove/nanda_experiment/powerlaw-1/initial_population.pkl.gz",
    compression="gzip",
)

data_f = pd.read_pickle(
    "../toremove/nanda_experiment/powerlaw-1/final_population.pkl.gz",
    compression="gzip",
)
data_f.head()
```

```python
P_i = data_i["P"]["[s]"].to_numpy()
P_dot_i = data_i["P_dot"]["[s s^-1]"].to_numpy()

P_f = data_f["P"]["[s]"].to_numpy()
P_dot_f = data_f["P_dot"]["[s s^-1]"].to_numpy()
L_radio_bol = data_f["L_radio_bol"]["[erg s^-1]"].to_numpy()
d = data_f["d"]["[kpc]"].to_numpy()
```

```python
mask_log_normal = (data_f['lognormal_flag'] == 1) 
mask_powerlaw = (data_f['lognormal_flag'] == 0) 
```

```python
fig, ax = plt.subplots(figsize=(12, 10))

ax.plot(
    P_i[np.transpose(mask_log_normal.to_numpy())[0]],
    P_dot_i[np.transpose(mask_log_normal.to_numpy())[0]],
    linestyle="None",
    marker="o",
    color="tab:cyan",
    markersize=1,
    alpha=0.2,
    rasterized=True,
    label = 'log-normal initial period'
)
ax.plot(
    P_i[np.transpose(mask_powerlaw.to_numpy())[0]],
    P_dot_i[np.transpose(mask_powerlaw.to_numpy())[0]],
    linestyle="None",
    marker="o",
    color="tab:green",
    markersize=1,
    alpha=0.2,
    rasterized=True,
    label = 'powerlaw initial period'
)


ax.plot(
    P_f[np.transpose(mask_log_normal.to_numpy())[0]],
    P_dot_f[np.transpose(mask_log_normal.to_numpy())[0]],
    linestyle="None",
    marker="o",
    color="tab:orange",
    markersize=1,
    alpha=0.2,
    rasterized=True,
    label = 'log-normal final period'

)

ax.plot(
    P_f[np.transpose(mask_powerlaw.to_numpy())[0]],
    P_dot_f[np.transpose(mask_powerlaw.to_numpy())[0]],
    linestyle="None",
    marker="o",
    color="tab:purple",
    markersize=1,
    alpha=0.2,
    rasterized=True,
    label = 'power-law final period'

)
#plt.axvline(10**(-0.7))
ax.set_xlabel(r"$P$ [s]")
ax.set_ylabel(r"$\dot{P}$")
plt.xscale('log') 
plt.yscale('log') 
plt.legend(bbox_to_anchor=(1, 1), frameon=False, loc=0, fontsize=20,markerscale=5, handler_map={PathCollection : HandlerPathCollection(update_func= update),
                        plt.Line2D : HandlerLine2D(update_func = update)})

#plt.savefig()
plt.show()
```

```python
P_bins = np.logspace(-3., 5., 31)

fig, ax = plt.subplots(figsize=(15,8))
hist_P_i, bin_edges_P_i = np.histogram(P_i, bins = P_bins)

ax.hist(
    P_i[np.transpose(mask_log_normal.to_numpy())[0]],
    bins=P_bins,
    histtype="step",
    color="tab:cyan",
    lw=4,
    ls=":",
    label='log-normal initial period',
    density=True,
)
ax.hist(
    P_i[np.transpose(mask_powerlaw.to_numpy())[0]],
    bins=P_bins,
    histtype="step",
    edgecolor="tab:green",
    lw=4,
    ls=":",
    label='powerlaw initial period',
    density=True,
)
ax.hist(
    P_f[np.transpose(mask_log_normal.to_numpy())[0]],
    bins=P_bins,
    histtype="step",
    edgecolor="tab:orange",
    lw=4,
    ls="-",
    label='log-normal final period',
    density=True,
)
ax.hist(
    P_f[np.transpose(mask_powerlaw.to_numpy())[0]],
    bins=P_bins,
    histtype="step",
    edgecolor="tab:purple",
    lw=4,
    ls="-",
    label='power-law final period',
    density=True,
)

plt.xlabel(r"$P$ [s]")
plt.ylabel(r"Number of NSs")
#plt.xlim(0., 50.0)
#plt.ylim(0.1, 2.e5)
plt.xscale('log')
plt.yscale('log')
plt.legend(bbox_to_anchor=(1.05, 1), frameon=False, loc=0, fontsize=20)

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(12, 10))

ax.plot(
    P_f[np.transpose(mask_log_normal.to_numpy())[0]],
    L_radio_bol[np.transpose(mask_log_normal.to_numpy())[0]],
    linestyle="None",
    marker="o",
    color="tab:orange",
    markersize=1,
    alpha=0.1,
    rasterized=True
)

ax.plot(
    P_f[np.transpose(mask_powerlaw.to_numpy())[0]],
    L_radio_bol[np.transpose(mask_powerlaw.to_numpy())[0]],
    linestyle="None",
    marker="o",
    color="tab:purple",
    markersize=1,
    alpha=0.1,
    rasterized=True
)

ax.set_xlabel(r"$P$ [s]")
ax.set_ylabel(r"$L_{\rm radio}$")
plt.xscale('log') 
plt.yscale('log') 

plt.show()
```
