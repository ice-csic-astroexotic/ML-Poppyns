# Comparing runs with best parameters

This notebook compares the birth rates and numbers of stars obtained for simulations with the best inferred parameters.

It provides the numerical values quoted in Section 5.6 of [Graber et al. (2024)](https://arxiv.org/abs/2312.14848).

Note that different numbers vary due to the stochastic nature of our simulation framework.

```python
import numpy as np
import pandas as pd
from scipy.stats import sem
```

## Birth rates

The birth rates for HTRU, PMPS and SMPS (in that order) as noted in the configuration file.

```python
parameters_run = np.zeros((10, 3))
```

```python
parameters_run[0] = np.array(
    [1.7, 2.04, 1.9]
)  # Celsa's initial run used for plots.
parameters_run[1] = np.array([1.6, 1.94, 1.975])
parameters_run[2] = np.array([1.6, 2.05, 1.8])
parameters_run[3] = np.array([1.6, 1.955, 1.82])
parameters_run[4] = np.array([1.7, 2.095, 1.8])
parameters_run[5] = np.array([1.6, 1.985, 1.8])
parameters_run[6] = np.array([1.7, 2.07, 1.9])
parameters_run[7] = np.array([1.7, 2.015, 1.9])
parameters_run[8] = np.array([1.7, 2.08, 1.95])
parameters_run[9] = np.array([1.7, 2.01, 1.6])
```

Extract individual rates.

```python
BR_HTRU = np.zeros(10)
BR_PMPS = np.zeros(10)
BR_SMPS = np.zeros(10)
```

```python
for i in range(len(BR_HTRU)):
    BR_HTRU[i] = parameters_run[i][0]
    BR_PMPS[i] = parameters_run[i][1]
    BR_SMPS[i] = parameters_run[i][2]
```

```python
print(BR_HTRU)
print(BR_PMPS)
print(BR_SMPS)
```

Means and standard errors of the means.

```python
print(f"PMPS birth rate mean: {np.round(np.mean(BR_PMPS), 2)}")
print(f"SMPS birth rate mean: {np.round(np.mean(BR_SMPS), 2)}")
print(f"HTRU birth rate mean: {np.round(np.mean(BR_HTRU), 2)}")
```

```python
sem_BR_HTRU = sem(BR_HTRU)
sem_BR_PMPS = sem(BR_PMPS)
sem_BR_SMPS = sem(BR_SMPS)
```

```python
print(f"PMPS birth rate standard error: {np.round(sem_BR_PMPS, 2)}")
print(f"SMPS birth rate standard error: {np.round(sem_BR_SMPS, 2)}")
print(f"HTRU birth rate standard error: {np.round(sem_BR_HTRU, 2)}")
```

## Population sizes

```python
counts_HTRU = np.zeros(10)
counts_PMPS = np.zeros(10)
counts_SMPS = np.zeros(10)
```

```python
for i in range(len(counts_HTRU)):
    # load individual pickle files for each survey
    df_PMPS_sim = pd.read_pickle(
        f"../../data/paper_results/graber_etal_2024/best_inferred_atnf/run_{i}/survey_PMPS_results.pkl.gz",
        compression="gzip",
    )

    df_SMPS_sim = pd.read_pickle(
        f"../../data/paper_results/graber_etal_2024/best_inferred_atnf/run_{i}/survey_SMPS_results.pkl.gz",
        compression="gzip",
    )

    df_HTRU_sim = pd.read_pickle(
        f"../../data/paper_results/graber_etal_2024/best_inferred_atnf/run_{i}/survey_HTRU_results_low_mid.pkl.gz",
        compression="gzip",
    )

    counts_HTRU[i] = len(df_HTRU_sim)
    counts_PMPS[i] = len(df_PMPS_sim)
    counts_SMPS[i] = len(df_SMPS_sim)
```

```python
print(counts_HTRU)
print(counts_PMPS)
print(counts_SMPS)
```

```python
print(f"PMPS counts mean: {np.round(np.mean(counts_PMPS), 0)}")
print(f"SMPS counts mean: {np.round(np.mean(counts_SMPS), 0)}")
print(f"HTRU counts mean: {np.round(np.mean(counts_HTRU), 0)}")
```

```python
sem_counts_HTRU = sem(counts_HTRU)
sem_counts_PMPS = sem(counts_PMPS)
sem_counts_SMPS = sem(counts_SMPS)
```

```python
print(f"PMPS counts standard error: {np.round(sem_counts_PMPS, 0)}")
print(f"SMPS counts standard error: {np.round(sem_counts_SMPS, 0)}")
print(f"HTRU counts standard error: {np.round(sem_counts_HTRU, 0)}")
```

Compare to observed numbers.

```python
observed_PMPS = 1009
observed_SMPS = 218
observed_HTRU = 1023
```

```python
fraction_PMPS = (
    100
    * np.abs(observed_PMPS - np.round(np.mean(counts_PMPS), 0))
    / observed_PMPS
)
fraction_SMPS = (
    100
    * np.abs(observed_SMPS - np.round(np.mean(counts_SMPS), 0))
    / observed_SMPS
)
fraction_HTRU = (
    100
    * np.abs(observed_HTRU - np.round(np.mean(counts_HTRU), 0))
    / observed_HTRU
)
```

```python
print(f"PMPS fraction: {np.round(fraction_PMPS, 0)}")
print(f"SMPS fraction: {np.round(fraction_SMPS, 0)}")
print(f"HTRU fraction: {np.round(fraction_HTRU, 0)}")
```
