# Train-validation losses for 1-parameter prediction
This notebook produces the train-validation losses curves for 1 parameter prediction (left plots in Fig. 6 in [Ronchi et al. 2021](https://ui.adsabs.harvard.edu/abs/2021ApJ...916..100R/abstract)).

```python
import json
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.ticker import ScalarFormatter

import utilities.plot_settings
```

```python
directory_path = "../../data/paper_results/ronchi_etal_2021/1par_train_valid_results/"

with open(f"{directory_path}hc_train_eval_result.json", "r") as read_file:
    train_data_hc = json.load(read_file)
with open(f"{directory_path}hc_validation_result.json", "r") as read_file:
    valid_data_hc = json.load(read_file)
with open(f"{directory_path}sigmak_train_eval_result.json", "r") as read_file:
    train_data_sigmak = json.load(read_file)
with open(f"{directory_path}sigmak_validation_result.json", "r") as read_file:
    valid_data_sigmak = json.load(read_file)
```

```python
epochs_hc = train_data_hc.keys()
epochs_sigmak = train_data_sigmak.keys()

train_loss_hc = []
valid_loss_hc = []

train_loss_sigmak = []
valid_loss_sigmak = []

for e in epochs_hc:
    train_loss_hc.append(train_data_hc[e]["h_c"])
    valid_loss_hc.append(valid_data_hc[e]["h_c"])
    
for e in epochs_sigmak:
    train_loss_sigmak.append(train_data_sigmak[e]["sigma_k"])
    valid_loss_sigmak.append(valid_data_sigmak[e]["sigma_k"])
        
epochs_hc = np.array(list(epochs_hc), dtype="float64") 
epochs_sigmak = np.array(list(epochs_sigmak), dtype="float64") 
```

```python
# plot loss trend h_c
colors = ["tab:red", "tab:blue", "tab:orange", "tab:green"]

fig, ax = plt.subplots(figsize=(15,9))

ax.plot(
    epochs_hc, 
    train_loss_hc, 
    linestyle="-",
    linewidth=3,
    color=colors[0],
    alpha=0.9,
    rasterized=True,
    label=r"Training" 
)
ax.plot(
    epochs_hc, 
    valid_loss_hc, 
    linestyle="-",
    linewidth=3,
    color=colors[1],
    alpha=0.9,
    rasterized=True,
    label=r"Validation" 
)

ax.set_xlabel(r'Training epoch')
ax.set_ylabel(r'$h_{\rm c}$ RMSE loss [kpc]')
ax.set_xscale('log')
#ax.set_yscale('log')
ax.xaxis.set_major_formatter(ScalarFormatter())
ax.legend(frameon=False, loc=0, fontsize=24)

plt.show()
```

```python
# plot loss trend sigma_k
fig, ax = plt.subplots(figsize=(15,9))

ax.plot(
    epochs_sigmak, 
    train_loss_sigmak, 
    linestyle="-",
    linewidth=3,
    color=colors[0],
    alpha=0.9,
    rasterized=True,
    label=r"Training" 
)
ax.plot(
    epochs_sigmak, 
    valid_loss_sigmak, 
    linestyle="-",
    linewidth=3,
    color=colors[1],
    alpha=0.9,
    rasterized=True,
    label=r"Validation" 
)

ax.set_xlabel(r'Training epoch')
ax.set_ylabel(r'$\sigma_{\rm k}$ RMSE loss [km s$^{-1}$]')
ax.set_xscale('log')
#ax.set_yscale('log')
ax.xaxis.set_major_formatter(ScalarFormatter())
ax.legend(frameon=False, loc=0, fontsize=24)

plt.show()
```
