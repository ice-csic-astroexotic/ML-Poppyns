# Data-representation and time-profiling tests for 2-parameter prediction
This notebook produces the plots in fig. 7 and 15 for the data-representation and time-profiling tests performed in [Ronchi et al. 2021](https://ui.adsabs.harvard.edu/abs/2021ApJ...916..100R/abstract)).

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import random
import time

from matplotlib.ticker import ScalarFormatter

import utilities.plot_settings
```

```python
batch_size = np.array([1, 4, 16, 64, 128])
samples = np.array([16, 64, 256, 1024, 4096])
```

```python
data = pd.read_csv('../../data/paper_results/ronchi_etal_2021/tests/maxwell_kick_plus_z_position_tests/gc_position_r128.txt', sep=" ")
gc_p_ba_sigmak = data["ba_sigmak"].to_numpy()
gc_p_ba_hc = data["ba_hc"].to_numpy()
gc_p_ft = data["FT"].to_numpy()
gc_p_bt = data["BT"].to_numpy()

gc_p_ft = gc_p_ft/batch_size
gc_p_bt = gc_p_bt/batch_size
```

```python
data = pd.read_csv('../../data/paper_results/ronchi_etal_2021/tests/maxwell_kick_plus_z_position_tests/gc_position_velocity_r128.txt', sep=" ")
gc_pv_ba_sigmak = data["ba_sigmak"].to_numpy()
gc_pv_ba_hc = data["ba_hc"].to_numpy()
gc_pv_ft = data["FT"].to_numpy()
gc_pv_bt = data["BT"].to_numpy()

gc_pv_ft = gc_pv_ft/batch_size
gc_pv_bt = gc_pv_bt/batch_size
```

```python
data = pd.read_csv('../../data/paper_results/ronchi_etal_2021/tests/maxwell_kick_plus_z_position_tests/icrs_position_r128.txt', sep=" ")
icrs_p_ba_sigmak = data["ba_sigmak"].to_numpy()
icrs_p_ba_hc = data["ba_hc"].to_numpy()
icrs_p_ft = data["FT"].to_numpy()
icrs_p_bt = data["BT"].to_numpy()

icrs_p_ft = icrs_p_ft/batch_size
icrs_p_bt = icrs_p_bt/batch_size
```

```python
data = pd.read_csv('../../data/paper_results/ronchi_etal_2021/tests/maxwell_kick_plus_z_position_tests/icrs_position_velocity_r128.txt', sep=" ")
icrs_pv_ba_sigmak = data["ba_sigmak"].to_numpy()
icrs_pv_ba_hc = data["ba_hc"].to_numpy()
icrs_pv_ft = data["FT"].to_numpy()
icrs_pv_bt = data["BT"].to_numpy()

icrs_pv_ft = icrs_pv_ft/batch_size
icrs_pv_bt = icrs_pv_bt/batch_size
```

## Figure 7

```python
colors = ["tab:red", "tab:blue", "tab:orange", "tab:green"]
```

```python
# plot best accuracies
fig, ax = plt.subplots()
#ax.title.set_text(r'$\sigma_{\rm k}$ accuracy')

ax.plot(
    samples, 
    gc_p_ba_sigmak, 
    linestyle=":",
    linewidth=5,
    color=colors[0],
    alpha=0.9,
    rasterized=True,
    label=r"GC pos" 
)
ax.plot(
    samples, 
    gc_pv_ba_sigmak, 
    linestyle="--",
    linewidth=5,
    color=colors[1],
    alpha=0.9,
    rasterized=True,
    label=r"GC pos+vel" 
)
ax.plot(
    samples, 
    icrs_p_ba_sigmak, 
    linestyle="-.",
    linewidth=5,
    color=colors[2],
    alpha=0.9,
    rasterized=True,
    label=r"ICRS pos" 
)
ax.plot(
    samples, 
    icrs_pv_ba_sigmak, 
    linestyle="-",
    linewidth=5,
    color=colors[3],
    alpha=0.9,
    rasterized=True,
    label=r"ICRS pos+vel" 
)

ax.set_xlabel(r'Sample number')
ax.set_ylabel(r'$\sigma_{\rm k}$ RMSE loss [km s$^{-1}$]')
ax.set_xscale('log')
ax.set_ylim(-0.5,18.)
ax.xaxis.set_major_formatter(ScalarFormatter())
ax.legend(frameon=False, loc=0, fontsize=22)

plt.savefig(
    f"plots/2par_CNN_sigmak_loss_test.pdf",
    dpi=300,
    bbox_inches="tight",
)
plt.show(block=False)
```

```python
fig, ax = plt.subplots()
#ax.title.set_text(r'$h_{\rm c}$ accuracy')

ax.plot(
    samples, 
    gc_p_ba_hc, 
    linestyle=":",
    linewidth=5,
    color=colors[0],
    alpha=0.9,
    rasterized=True,
    label=r"GC pos" 
)
ax.plot(
    samples, 
    gc_pv_ba_hc, 
    linestyle="--",
    linewidth=5,
    color=colors[1],
    alpha=0.9,
    rasterized=True,
    label=r"GC pos+vel" 
)
ax.plot(
    samples, 
    icrs_p_ba_hc, 
    linestyle="-.",
    linewidth=5,
    color=colors[2],
    alpha=0.9,
    rasterized=True,
    label=r"ICRS pos" 
)
ax.plot(
    samples, 
    icrs_pv_ba_hc, 
    linestyle="-",
    linewidth=5,
    color=colors[3],
    alpha=0.9,
    rasterized=True,
    label=r"ICRS pos+vel" 
)

ax.set_xlabel(r'Sample number')
ax.set_ylabel(r'$h_{\rm c}$ RMSE loss [kpc]')
ax.set_xscale('log')
ax.set_ylim(0.,0.65)
ax.xaxis.set_major_formatter(ScalarFormatter())
ax.legend(frameon=False, loc=0, fontsize=22)

plt.savefig(
    f"plots/2par_CNN_hc_loss_test.pdf",
    dpi=300,
    bbox_inches="tight",
)
plt.show(block=False)
```

## Figure 15

```python
# plot forward and backward timing 

fig, ax = plt.subplots()

ax.plot(
    batch_size, 
    gc_p_ft, 
    linestyle="-",
    linewidth=5,
    color=colors[0],
    alpha=0.9,
    rasterized=True,
    label=r"GC pos" 
)
ax.plot(
    batch_size,
    gc_p_bt, 
    linestyle="--",
    linewidth=5,
    color=colors[0],
    alpha=0.9,
    rasterized=True,
)
ax.plot(
    batch_size, 
    gc_pv_ft, 
    linestyle="-",
    linewidth=5,
    color=colors[1],
    alpha=0.9,
    rasterized=True,
    label=r"GC pos+vel" 
)
ax.plot(
    batch_size,
    gc_pv_bt, 
    linestyle="--",
    linewidth=5,
    color=colors[1],
    alpha=0.9,
    rasterized=True,
)
ax.plot(
    batch_size, 
    icrs_p_ft, 
    linestyle="-",
    linewidth=5,
    color=colors[2],
    alpha=0.9,
    rasterized=True,
    label=r"ICRS pos" 
)
ax.plot(
    batch_size,
    icrs_p_bt, 
    linestyle="--",
    linewidth=5,
    color=colors[2],
    alpha=0.9,
    rasterized=True,
)
ax.plot(
    batch_size, 
    icrs_pv_ft, 
    linestyle="-",
    linewidth=5,
    color=colors[3],
    alpha=0.9,
    rasterized=True,
    label=r"ICRS pos+vel" 
)
ax.plot(
    batch_size,
    icrs_pv_bt, 
    linestyle="--",
    linewidth=5,
    color=colors[3],
    alpha=0.9,
    rasterized=True,
)

ax.set_xlabel(r'Batch size')
ax.set_ylabel(r'Time [ms]')
ax.set_xscale('log')
ax.set_yscale('log')
for axis in [ax.xaxis, ax.yaxis]:
    axis.set_major_formatter(ScalarFormatter())
ax.legend(frameon=False, loc=0, fontsize=22)

plt.savefig(
    f"plots/2par_CNN_time_test.pdf",
    dpi=300,
    bbox_inches="tight",
)
plt.show(block=False)
```
