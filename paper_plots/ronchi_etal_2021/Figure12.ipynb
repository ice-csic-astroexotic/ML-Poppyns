# Figure 12
This notebook reproduce Fig. 12 in [Ronchi et al. 2021](https://ui.adsabs.harvard.edu/abs/2021ApJ...916..100R/abstract).

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import random

import utilities.plot_settings
```

```python
# plot RMSE validation accuracy and MRE as a function of sampled stars from the simulations
n_stars = np.array([200,400,600,800,1000,1200,1400,1600,1800,2000])
rmse_hc = np.zeros(len(n_stars))
rmse_sigmak = np.zeros(len(n_stars))
mre_hc = np.zeros(len(n_stars))
mre_sigmak = np.zeros(len(n_stars))

i = 0
for n in n_stars:
    
    data = pd.read_csv(f"../../data/paper_results/ronchi_etal_2021/observation_comparison/{n}stars_inference_results.csv")
    target_hc = data["target:h_c"].to_numpy()
    target_sigmak = data["target:sigma_k"].to_numpy()
    prediction_hc = data["predicted:h_c"].to_numpy()
    prediction_sigmak = data["predicted:sigma_k"].to_numpy()
       
    rmse_hc[i] = np.sqrt(np.mean((prediction_hc - target_hc)**2))
    rmse_sigmak[i] = np.sqrt(np.mean((prediction_sigmak - target_sigmak)**2))
    
    mre_hc[i] = np.mean(abs((prediction_hc - target_hc)/target_hc))
    mre_sigmak[i] = np.mean(abs((prediction_sigmak - target_sigmak)/target_sigmak))
    
    i = i+1
```

```python
fig, ax1 = plt.subplots(figsize=(15,9))

ax1.set_xlabel(r'Number of NSs')
ax1.set_ylabel(r'$\sigma_{\rm k}$ RMSE [km s$^{-1}$]')
ax1.plot(
    n_stars,
    rmse_sigmak,
    linestyle="-",
    linewidth=3,
    marker="^",
    color="tab:red",
    markersize=10,
    alpha=1,
    rasterized=True,
    label=r"$\sigma_{\rm k}$",
)

ax2 = ax1.twinx()  # instantiate a second axes that shares the same x-axis

ax2.set_ylabel(r'$h_{\rm c}$ RMSE [kpc]')
ax2.plot(
    n_stars,
    rmse_hc,
    linestyle="-",
    linewidth=3,
    marker="s",
    color="tab:blue",
    markersize=10,
    alpha=1,
    rasterized=True,
    label=r"$h_{\rm c}$",
)
# ask matplotlib for the plotted objects and their labels
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax2.legend(lines1 + lines2, labels1 + labels2, loc=1, fontsize=24, frameon=False)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(15,9))
ax.set_xlabel(r'Number of NSs')
ax.set_ylabel(r'MRE')
ax.plot(
    n_stars,
    mre_sigmak,
    linestyle="-",
    linewidth=3,
    marker="^",
    color="tab:red",
    markersize=10,
    alpha=1,
    rasterized=True,
    label=r"$\sigma_{\rm k}$",
)
ax.plot(
    n_stars,
    mre_hc,
    linestyle="-",
    linewidth=3,
    marker="s",
    color="tab:blue",
    markersize=10,
    alpha=1,
    rasterized=True,
    label=r"$h_{\rm c}$",
)
ax.legend(frameon=False, loc=1, fontsize=24)
plt.show()
```
