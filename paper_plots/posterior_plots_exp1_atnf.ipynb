# Two-dimensional marginalised posteriors

This notebook reproduces Figures 8 and 9 in [Graber et al. (2024)](https://arxiv.org/abs/2312.14848) and shows 2D marginalised posteriors in the form of corner plots.

```python
import corner
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import torch
```

```python
from matplotlib import rcParams
from matplotlib import rc
import matplotlib as mpl

rc("text", usetex=True)
rc("font", family="serif")
mpl.rcParams["text.latex.preamble"] = r"\usepackage{amsmath}"
```

```python
SMALL_SIZE = 14
MEDIUM_SIZE = 16
BIGGER_SIZE = 36

plt.rc("font", size=SMALL_SIZE)  # controls default text sizes
plt.rc("axes", titlesize=MEDIUM_SIZE)  # fontsize of the axes title
plt.rc("axes", labelsize=MEDIUM_SIZE)  # fontsize of the x and y labels
plt.rc("xtick", labelsize=SMALL_SIZE)  # fontsize of the tick labels
plt.rc("ytick", labelsize=SMALL_SIZE)  # fontsize of the tick labels
plt.rc("legend", fontsize=SMALL_SIZE)  # legend fontsize
plt.rc("figure", titlesize=BIGGER_SIZE)  # fontsize of the figure title
```

```python
plot_titles_corner = False
# format = "png"
format = "pdf"
```

Load PyTorch posterior samples (1D marginalised samples).

```python
samples_test_1 = (
    torch.load("../paper_plots/samples/samples_exp_1_test_1.pt")
    .detach()
    .cpu()
    .numpy()
)
samples_test_2 = (
    torch.load("../paper_plots/samples/samples_exp_1_test_2.pt")
    .detach()
    .cpu()
    .numpy()
)
samples_test_3 = (
    torch.load("../paper_plots/samples/samples_exp_1_test_3.pt")
    .detach()
    .cpu()
    .numpy()
)
samples_atnf = (
    torch.load("../paper_plots/samples/samples_exp_1_atnf.pt")
    .detach()
    .cpu()
    .numpy()
)
```

Load ground truths.

```python
dataset = pd.read_csv(
    "../paper_plots/samples/dataset_test_cornerplot_res32.csv"
)
```

```python
ground_truth_1 = dataset.iloc[0][
    [
        "B_initial_log10_mean",
        "B_initial_log10_sigma",
        "P_initial_log10_mean",
        "P_initial_log10_sigma",
        "a_late",
    ]
].values
ground_truth_2 = dataset.iloc[1][
    [
        "B_initial_log10_mean",
        "B_initial_log10_sigma",
        "P_initial_log10_mean",
        "P_initial_log10_sigma",
        "a_late",
    ]
].values
ground_truth_3 = dataset.iloc[2][
    [
        "B_initial_log10_mean",
        "B_initial_log10_sigma",
        "P_initial_log10_mean",
        "P_initial_log10_sigma",
        "a_late",
    ]
].values
```

```python
print(ground_truth_1)
print(ground_truth_2)
print(ground_truth_3)
```

Determining the corresponding medians.

```python
median_1 = np.median(samples_test_1, axis=0)
median_2 = np.median(samples_test_2, axis=0)
median_3 = np.median(samples_test_3, axis=0)
median_atnf = np.median(samples_atnf, axis=0)

print(np.round_(median_1, 2))
print(np.round_(median_2, 2))
print(np.round_(median_3, 2))
print(np.round_(median_atnf, 2))
```

```python
colors = ["#fde725", "dodgerblue", "#440154"]
```

## Corner plots

```python
limits = [[12.0, 14.0], [0.1, 1.0], [-1.5, -0.3], [0.1, 1.0], [-3, -0.5]]

parameter_labels = [
    r"$\mu_{\log B}$",
    r"$\sigma_{\log B}$",
    r"$\mu_{\log P}$",
    r"$\sigma_{\log P}$",
    r"$a_{\rm late}$",
]
```

```python
# Corner plot of the inferred posterior distributions for each parameter.
figure = corner.corner(
    samples_test_1,
    bins=32,
    labels=parameter_labels,
    label_kwargs={"fontsize": BIGGER_SIZE},
    range=limits,
    quantiles=[0.025, 0.5, 0.975],
    levels=(
        1 - np.exp(-0.5),
        1 - np.exp(-2),
        1 - np.exp(-9.0 / 2.0),
    ),  # 1, 2 and 3 sigma levels
    show_titles=plot_titles_corner,
    title_kwargs={"fontsize": MEDIUM_SIZE},
)

corner.overplot_lines(figure, ground_truth_1, color=colors[1])
corner.overplot_lines(figure, median_1, color=colors[2])

corner.overplot_points(
    figure,
    ground_truth_1[None],
    marker="o",
    color=colors[1],
)
corner.overplot_points(
    figure,
    median_1[None],
    marker="o",
    color=colors[2],
)

figure.suptitle(r"Test sample 1")

plt.tight_layout()
plt.savefig(
    f"../paper_plots/plots/inference_exp_1_test_1.{format}",
    dpi=400,
    bbox_inches="tight",
)
plt.show()
```

```python
figure = corner.corner(
    samples_test_2,
    bins=32,
    labels=parameter_labels,
    label_kwargs={"fontsize": BIGGER_SIZE},
    range=limits,
    quantiles=[0.025, 0.5, 0.975],
    levels=(
        1 - np.exp(-0.5),
        1 - np.exp(-2),
        1 - np.exp(-9.0 / 2.0),
    ),  # 1, 2 and 3 sigma levels
    show_titles=plot_titles_corner,
    title_kwargs={"fontsize": MEDIUM_SIZE},
)

corner.overplot_lines(figure, ground_truth_2, color=colors[1])
corner.overplot_lines(figure, median_2, color=colors[2])

corner.overplot_points(
    figure,
    ground_truth_2[None],
    marker="o",
    color=colors[1],
)
corner.overplot_points(
    figure,
    median_2[None],
    marker="o",
    color=colors[2],
)

figure.suptitle(r"Test sample 2")

plt.tight_layout()
plt.savefig(
    f"../paper_plots/plots/inference_exp_1_test_2.{format}",
    dpi=400,
    bbox_inches="tight",
)
plt.show()
```

```python
# Corner plot of the inferred posterior distributions for each parameter.
figure = corner.corner(
    samples_test_3,
    bins=32,
    labels=parameter_labels,
    label_kwargs={"fontsize": BIGGER_SIZE},
    range=limits,
    quantiles=[0.025, 0.5, 0.975],
    levels=(
        1 - np.exp(-0.5),
        1 - np.exp(-2),
        1 - np.exp(-9.0 / 2.0),
    ),  # 1, 2 and 3 sigma levels
    show_titles=plot_titles_corner,
    title_kwargs={"fontsize": MEDIUM_SIZE},
)

corner.overplot_lines(figure, ground_truth_3, color=colors[1])
corner.overplot_lines(figure, median_3, color=colors[2])

corner.overplot_points(
    figure,
    ground_truth_3[None],
    marker="o",
    color=colors[1],
)
corner.overplot_points(
    figure,
    median_3[None],
    marker="o",
    color=colors[2],
)

figure.suptitle(r"Test sample 3")

plt.tight_layout()
plt.savefig(
    f"../paper_plots/plots/inference_exp_1_test_3.{format}",
    dpi=400,
    bbox_inches="tight",
)
plt.show()
```

## Medians and confidence intervals

```python
quantiles_test_1 = np.quantile(samples_test_1.T, [0.025, 0.5, 0.975], axis=1)
quantiles_test_2 = np.quantile(samples_test_2.T, [0.025, 0.5, 0.975], axis=1)
quantiles_test_3 = np.quantile(samples_test_3.T, [0.025, 0.5, 0.975], axis=1)
quantiles_atnf = np.quantile(samples_atnf.T, [0.025, 0.5, 0.975], axis=1)
```
