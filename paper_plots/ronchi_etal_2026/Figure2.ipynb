# Figure 2

In this notebook, we will plot eight density maps for the observed population and produce Figure 2 of Ronchi et al. (2026): four $P$-$\dot{P}$ maps and four flux-averaged maps displayed in $P$-$\dot{P}$ space. In the former, the color represents the density in number of neutron stars in each bin, while in the latter, it represents the average flux per bin.

```python
import numpy as np
import matplotlib.pyplot as plt

import mlpoppyns.generator.maps.maps_utils as mu
import utilities.plot_settings

from matplotlib import rc

rc("text", usetex=True)
rc("font", family="serif")
plt.rcParams["text.latex.preamble"] = r"\usepackage{amsmath}"
```

```python
SMALL_SIZE = 20
MEDIUM_SIZE = 26
BIGGER_SIZE = 26

plt.rc("font", size=SMALL_SIZE)  # controls default text sizes
plt.rc("axes", titlesize=BIGGER_SIZE)  # fontsize of the axes title
plt.rc("axes", labelsize=BIGGER_SIZE)  # fontsize of the x and y labels
plt.rc("xtick", labelsize=MEDIUM_SIZE)  # fontsize of the tick labels
plt.rc("ytick", labelsize=MEDIUM_SIZE)  # fontsize of the tick labels
plt.rc("legend", fontsize=SMALL_SIZE)  # legend fontsize
plt.rc("figure", titlesize=BIGGER_SIZE)  # fontsize of the figure title
```

```python
resolution = 32
```

```python
n_x_bins = resolution
n_y_bins = resolution

x_range = (0.01, 100.0)
y_range = (1.0e-20, 1.0e-9)

x_log_scale = True
y_log_scale = True

_, _, x_edges, y_edges = mu.log_scale_vs_linear_scale(
    np.ones(5),
    np.ones(5),
    x_range,
    y_range,
    x_log_scale,
    y_log_scale,
    n_x_bins,
    n_y_bins,
)

colormap = "viridis"
colormap_fluxes = "cividis"
```

```python
import os

print(os.getcwd())
```

Loading the observed sample that we will use to generate the density maps.

```python
directory_path = "../../data/example_generator_observed"
density_ppdot_PMPS = np.load(
    directory_path + "/survey_PMPS_density_map_ppdot_0.npy"
)
density_ppdot_SMPS = np.load(
    directory_path + "/survey_SMPS_density_map_ppdot_0.npy"
)
density_ppdot_HTRU = np.load(
    directory_path + "/survey_HTRU_density_map_ppdot_0.npy"
)
density_ppdot_xray = np.load(
    directory_path + "/survey_xray_density_map_ppdot_0.npy"
)
density_ppdot_fluxes_PMPS = np.load(
    directory_path + "/survey_PMPS_flux_map_ppdot_0.npy"
)
density_ppdot_fluxes_SMPS = np.load(
    directory_path + "/survey_SMPS_flux_map_ppdot_0.npy"
)
density_ppdot_fluxes_HTRU = np.load(
    directory_path + "/survey_HTRU_flux_map_ppdot_0.npy"
)
density_ppdot_fluxes_xray = np.load(
    directory_path + "/survey_xray_flux_map_ppdot_0.npy"
)
# Check resolution is correct.
print(np.shape(density_ppdot_PMPS))
print(np.shape(density_ppdot_SMPS))
print(np.shape(density_ppdot_HTRU))
print(np.shape(density_ppdot_xray))
```

```python
datasets = [
    density_ppdot_PMPS.T,
    density_ppdot_SMPS.T,
    density_ppdot_HTRU.T,
    density_ppdot_xray.T,
    density_ppdot_fluxes_PMPS.T,
    density_ppdot_fluxes_SMPS.T,
    density_ppdot_fluxes_HTRU.T,
    density_ppdot_fluxes_xray.T,
]
colormaps = [
    colormap,
    colormap,
    colormap,
    colormap,
    colormap_fluxes,
    colormap_fluxes,
    colormap_fluxes,
    colormap_fluxes,
]

titles = [
    r"$P\dot{P}$ density map: PMPS",
    r"$P\dot{P}$ density map: SMPS",
    r"$P\dot{P}$ density map: HTRU",
    r"$P\dot{P}$ density map: X-ray"
    r"Fluxes displayed in $P\dot{P}$ space: PMPS",
    r"Fluxes displayed in $P\dot{P}$ space: SMPS",
    r"Fluxes displayed in $P\dot{P}$ space: HTRU",
    r"Fluxes displayed in $P\dot{P}$ space: X-ray",
]
```

```python
for i in range(len(datasets)):
    fig = plt.figure(frameon=False)
    
    fig.set_size_inches(n_x_bins / 5, n_y_bins / 5)
    ax = plt.Axes(fig, [0.0, 0.0, 1.0, 1.0])
    ax.set_axis_off()
    fig.add_axes(ax)
    
    # Generating a pseudocolor plot of the smeared out density distribution.
    # We transpose the array as pcolormesh is indexed starting from the lower
    # left , i.e., the column (row) index corresponds to the x (y) coordinate.
    
    ax.pcolormesh(x_edges, y_edges, datasets[i], cmap=colormaps[i])
    # ax.set_xlim(x_range[0], x_range[1])
    # ax.set_ylim(y_range[0], y_range[1])
    ax.set_title(titles[i], fontsize=MEDIUM_SIZE)
    
    plt.show()
```

Save all maps in a single figure.

```python
fig, axes = plt.subplots(2, 4, figsize=(15, 8), constrained_layout=True)

# Axes for shared labels.
x_label = r"$P$ [s]"
y_label = r"$\dot{P}$ [s/s]"

# Plot each map.
for i, (data, ax) in enumerate(zip(datasets, axes.flat)):
    # Plot the data
    psm = ax.pcolormesh(x_edges, y_edges, data, cmap=colormaps[i])

    # Set tick direction outside for the leftmost and bottom panels and remove those ticks that are not needed.
    if i % 4 == 0:  # Leftmost column.
        ax.set_ylabel(y_label)
        ax.tick_params(
            axis="y", direction="out", labelright=False, right=False, top=False
        )
    else:
        ax.tick_params(
            labelleft=False,
            direction="out",
            labelright=False,
            right=False,
            left=False,
            top=False,
        )

    if i // 4 == 1:  # Bottom row.
        ax.set_xlabel(x_label)
        ax.tick_params(axis="x", direction="out", labelright=False, top=False)
    else:
        ax.tick_params(
            labelbottom=False, bottom=False, labelright=False, top=False
        )

plt.savefig("plots/merged_input_maps.pdf", dpi=400, bbox_inches="tight")
plt.show()
```
