# Input maps

In this notebook, we will plot six density maps for a given test sample in Figure 1 of Pardo-Araujo et al 2025: three $P$-$\dot{P}$ maps and three flux-averaged maps displayed in $P$-$\dot{P}$ space maps. In the former, the color represents the density in number of pulsars in each bin, while in the latter, it represents the average flux per bin.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import rcParams
from matplotlib import rc
from scipy.integrate import quad
from scipy import stats
from matplotlib.gridspec import GridSpec

from pypopsyn.simulator.config_simulator import cfg
import pypopsyn.simulator.basics.constants as const
import pypopsyn.generator.maps.axes_scaling as axs
import utilities.plot_settings

from matplotlib import rc
import pypopsyn.learning.loaders.loader_multichannel_array as dl


rc("text", usetex=True)
rc("font", family="serif")
plt.rcParams["text.latex.preamble"] = r"\usepackage{amsmath}"
```

```python
SMALL_SIZE = 20
MEDIUM_SIZE = 26
BIGGER_SIZE = 26

plt.rc("font", size=SMALL_SIZE)  # controls default text sizes
plt.rc("axes", titlesize=BIGGER_SIZE)  # fontsize of the axes title
plt.rc("axes", labelsize=BIGGER_SIZE)  # fontsize of the x and y labels
plt.rc("xtick", labelsize=MEDIUM_SIZE)  # fontsize of the tick labels
plt.rc("ytick", labelsize=MEDIUM_SIZE)  # fontsize of the tick labels
plt.rc("legend", fontsize=SMALL_SIZE)  # legend fontsize
plt.rc("figure", titlesize=BIGGER_SIZE)  # fontsize of the figure title
```

```python
resolution = 32
```

```python
n_x_bins = resolution
n_y_bins = resolution

x_range = (0.001, 100.0)
y_range = (1.0e-21, 1.0e-9)

x_log_scale = True
y_log_scale = True

x_edges, y_edges = axs.log_scale_vs_linear_scale(
    x_range,
    y_range,
    x_log_scale,
    y_log_scale,
    n_x_bins,
    n_y_bins,
)

colormap = "viridis"
colormap_fluxes = "cividis"
```

Loading the test sample that we will use to generate the density maps.

```python
directory_path = "../../data/paper_results/pardo_etal_2025/best_inferred_atnf_density_maps"
density_ppdot_PMPS = np.load(
    directory_path + "/survey_PMPS_ppdot_map_0.npy"
)
density_ppdot_SMPS = np.load(
    directory_path + "/survey_SMPS_ppdot_map_0.npy"
)
density_ppdot_HTRU = np.load(
    directory_path + "/survey_HTRU_ppdot_map_0.npy"
)
density_ppdot_fluxes_PMPS = np.load(
    directory_path + "/survey_PMPS_ppdot_map_fluxes_0.npy"
)
density_ppdot_fluxes_SMPS = np.load(
    directory_path + "/survey_SMPS_ppdot_map_fluxes_0.npy"
)
density_ppdot_fluxes_HTRU = np.load(
    directory_path + "/survey_HTRU_ppdot_map_fluxes_0.npy"
)
# Check resolution is correct.
print(np.shape(density_ppdot_PMPS))
print(np.shape(density_ppdot_SMPS))
print(np.shape(density_ppdot_HTRU))
```

```python
fig = plt.figure(frameon=False)

fig.set_size_inches(n_x_bins / 5, n_y_bins / 5)
ax = plt.Axes(fig, [0.0, 0.0, 1.0, 1.0])
ax.set_axis_off()
fig.add_axes(ax)

# Generating a pseudocolor plot of the smeared out density distribution;
# we transpose the array as pcolormesh is indexed starting from the lower
# left , i.e., the column (row) index corresponds to the x (y) coordinate.
ax.pcolormesh(x_edges, y_edges, density_ppdot_PMPS.T, cmap=colormap)
ax.set_xlim(x_range[0], x_range[1])
ax.set_ylim(y_range[0], y_range[1])
ax.set_title(r"$P\dot{P}$ density map: PMPS", fontsize=MEDIUM_SIZE)

if x_log_scale:
    ax.set_xscale("log")

if y_log_scale:
    ax.set_yscale("log")

plt.savefig(
    f"../../paper_plots/pardo_et_al_2025/plots/simulated_ppdot_PMPS.pdf",
    dpi=400,
    bbox_inches="tight",
)
plt.show()
```

Note that the shapes in both plots do not coincide because the fluxes are taken from MeerKAT, while the PMPS, SMPS, and HTRU surveys are incomplete for this dataset. As a result, we need to select a subset from each of these surveys before plotting the fluxes. The size of each subset is calculated in the `observed_population.ipynb` notebook and corresponds to the overlapping pulsars between MeerKAT and each survey.

```python
fig, axes = plt.subplots(2, 3, figsize=(15, 10), constrained_layout=True)

datasets = [
    density_ppdot_PMPS.T, density_ppdot_SMPS.T, density_ppdot_HTRU.T,
    density_ppdot_fluxes_PMPS.T, density_ppdot_fluxes_SMPS.T, density_ppdot_fluxes_HTRU.T
]
colormaps = [colormap, colormap, colormap, colormap_fluxes, colormap_fluxes, colormap_fluxes]

# Axes for shared labels.
x_label = r"$P$ [s]"
y_label = r"$\dot{P}$ [s/s]"
psm = []
# Plot each map.
for i, (data, ax) in enumerate(zip(datasets, axes.flat)):
    # Plot the data
    psm.append(ax.pcolormesh(x_edges, y_edges, data, cmap=colormaps[i]))
    ax.set_xlim(x_range[0], x_range[1])
    ax.set_ylim(y_range[0], y_range[1])

    ax.set_xscale("log")
    ax.set_yscale("log")

    # Set tick direction outside for the leftmost and bottom panels and remove those ticks that are not needed.
    if i % 3 == 0:  # Leftmost column.
        ax.set_ylabel(y_label)
        ax.tick_params(axis='y', direction='out', labelright=False, right=False, top=False)
    else:
        ax.tick_params(labelleft=False, direction='out', labelright=False, right=False, left=False, top=False)

    if i // 3 == 1:  # Bottom row.
        ax.set_xlabel(x_label)
        ax.tick_params(axis='x', direction='out', labelright=False, top=False)
    else:
        ax.tick_params(labelbottom=False, bottom=False, labelright=False, top=False)

# Add a colorbar for the first row (last plot in first row is index 2)
cbar1 = fig.colorbar(psm[2], ax=axes[0, 2], orientation='vertical')
cbar1.set_label('NS counts')

# Add a colorbar for the second row (last plot in second row is index 5)
cbar2 = fig.colorbar(psm[5], ax=axes[1, 2], orientation='vertical')
cbar2.set_label(r"log$_{10}$ $S_{{\rm mean}, 1400}$ [Jy]")

plt.savefig("../../paper_plots/pardo_et_al_2025/plots/merged_input_maps.pdf", dpi=400, bbox_inches="tight")
plt.show()
```
