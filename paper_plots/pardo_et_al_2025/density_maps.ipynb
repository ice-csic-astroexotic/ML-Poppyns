```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib import rcParams
from matplotlib import rc
import matplotlib as mpl
from scipy.integrate import quad
from scipy import stats

from pypopsyn.simulator.config_simulator import cfg
import pypopsyn.simulator.basics.constants as const
import pypopsyn.generator.maps.axes_scaling as axs
import utilities.plot_settings

from matplotlib import rc
import pypopsyn.learning.loaders.loader_multichannel_array_stat as dl


rc("text", usetex=True)
rc("font", family="serif")
mpl.rcParams["text.latex.preamble"] = r"\usepackage{amsmath}"
```

```python
SMALL_SIZE = 30
MEDIUM_SIZE = 40
BIGGER_SIZE = 50

plt.rc("font", size=SMALL_SIZE)  # controls default text sizes
plt.rc("axes", titlesize=BIGGER_SIZE)  # fontsize of the axes title
plt.rc("axes", labelsize=BIGGER_SIZE)  # fontsize of the x and y labels
plt.rc("xtick", labelsize=MEDIUM_SIZE)  # fontsize of the tick labels
plt.rc("ytick", labelsize=MEDIUM_SIZE)  # fontsize of the tick labels
plt.rc("legend", fontsize=SMALL_SIZE)  # legend fontsize
plt.rc("figure", titlesize=BIGGER_SIZE)  # fontsize of the figure title
```

```python
resolution = 32
```

```python
n_x_bins = resolution
n_y_bins = resolution

x_range = (0.001, 100.0)
y_range = (1.0e-21, 1.0e-9)

x_log_scale = True
y_log_scale = True

x_edges, y_edges = axs.log_scale_vs_linear_scale(
    x_range,
    y_range,
    x_log_scale,
    y_log_scale,
    n_x_bins,
    n_y_bins,
)

colormap = "viridis"
print(y_edges)
```

```python
density_ppdot_PMPS = np.load(
    "../../data/paper_results/pardo_etal_2025/test_simulation/survey_PMPS_ppdot_map_0.npy"
)
density_ppdot_SMPS = np.load(
    "../../data/paper_results/pardo_etal_2025/test_simulation/survey_SMPS_ppdot_map_0.npy"
)
density_ppdot_HTRU = np.load(
    "../../data/paper_results/pardo_etal_2025/test_simulation/survey_HTRU_ppdot_map_0.npy"
)
density_ppdot_fluxes_PMPS = np.load(
    "../../data/paper_results/pardo_etal_2025/test_simulation/survey_PMPS_ppdot_map_fluxes_0.npy"
)
density_ppdot_fluxes_SMPS = np.load(
    "../../data/paper_results/pardo_etal_2025/test_simulation/survey_SMPS_ppdot_map_fluxes_0.npy"
)
density_ppdot_fluxes_HTRU = np.load(
    "../../data/paper_results/pardo_etal_2025/test_simulation/survey_HTRU_ppdot_map_fluxes_0.npy"
)
# Check resolution is correct
print(np.shape(density_ppdot_PMPS))
print(np.shape(density_ppdot_SMPS))
print(np.shape(density_ppdot_HTRU))
```

```python
fig = plt.figure(frameon=False)

fig.set_size_inches(n_x_bins / 5, n_y_bins / 5)
ax = plt.Axes(fig, [0.0, 0.0, 1.0, 1.0])
ax.set_axis_off()
fig.add_axes(ax)

# Generating a pseudocolor plot of the smeared out density distribution;
# we transpose the array as pcolormesh is indexed starting from the lower
# left , i.e., the column (row) index corresponds to the x (y) coordinate.
ax.pcolormesh(x_edges, y_edges, density_ppdot_PMPS.T, cmap=colormap)
ax.set_xlim(x_range[0], x_range[1])
ax.set_ylim(y_range[0], y_range[1])
ax.set_title(f"PPdot density map: PMPS", fontsize=MEDIUM_SIZE)

if x_log_scale:
    ax.set_xscale("log")

if y_log_scale:
    ax.set_yscale("log")

plt.savefig(
    f"../../paper_plots/pardo_et_al_2025/plots/simulated_ppdot_PMPS.pdf",
    dpi=400,
    bbox_inches="tight",
)
plt.show()
```
