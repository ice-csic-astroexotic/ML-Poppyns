### Loading the data

```python
df_PMPS_op1 = []
df_SMPS_op1 = []
df_HTRU_low_mid_op1 = []

df_SKA_low_AA4_op1 = []
df_SKA_mid_band1_AA4_op1 = []
df_SKA_mid_band2_AA4_op1 = []

df_SKA_low_AAstar_op1 = []
df_SKA_mid_band1_AAstar_op1 = []
df_SKA_mid_band2_AAstar_op1 = []

path = "../../SKA_census_simulations_si145/survey_option1"

for i in range(1, run_no + 1):
    df_PMPS_op1.append(
        pd.read_pickle(
            f"{path}/run_{i}/survey_PMPS_results.pkl.gz", compression="gzip"
        )
    )
    df_SMPS_op1.append(
        pd.read_pickle(
            f"{path}/run_{i}/survey_SMPS_results.pkl.gz", compression="gzip"
        )
    )
    df_HTRU_low_mid_op1.append(
        pd.read_pickle(
            f"{path}/run_{i}/survey_HTRU_low_mid_results.pkl.gz",
            compression="gzip",
        )
    )

    df_SKA_low_AA4_op1.append(
        pd.read_pickle(
            f"{path}/run_{i}/survey_SKA_low_AA4_results.pkl.gz",
            compression="gzip",
        )
    )
    df_SKA_mid_band1_AA4_op1.append(
        pd.read_pickle(
            f"{path}/run_{i}/survey_SKA_mid_band1_AA4_results.pkl.gz",
            compression="gzip",
        )
    )
    df_SKA_mid_band2_AA4_op1.append(
        pd.read_pickle(
            f"{path}/run_{i}/survey_SKA_mid_band2_AA4_results.pkl.gz",
            compression="gzip",
        )
    )

    df_SKA_low_AAstar_op1.append(
        pd.read_pickle(
            f"{path}/run_{i}/survey_SKA_low_AAstar_results.pkl.gz",
            compression="gzip",
        )
    )
    df_SKA_mid_band1_AAstar_op1.append(
        pd.read_pickle(
            f"{path}/run_{i}/survey_SKA_mid_band1_AAstar_results.pkl.gz",
            compression="gzip",
        )
    )
    df_SKA_mid_band2_AAstar_op1.append(
        pd.read_pickle(
            f"{path}/run_{i}/survey_SKA_mid_band2_AAstar_results.pkl.gz",
            compression="gzip",
        )
    )
```

### Survey count with a death line

Implementation of death line based on a twisted, multi-polar magnetic field.

Extracting periods and period derivatives:

```python
periods_PMPS_op1 = []
periods_SMPS_op1 = []
periods_HTRU_low_mid_op1 = []

periods_SKA_low_AA4_op1 = []
periods_SKA_mid_band1_AA4_op1 = []
periods_SKA_mid_band2_AA4_op1 = []

periods_SKA_low_AAstar_op1 = []
periods_SKA_mid_band1_AAstar_op1 = []
periods_SKA_mid_band2_AAstar_op1 = []

for i in range(run_no):
    periods_PMPS_op1.append(
        df_PMPS_op1[i]["P"]["[s]"].to_numpy().astype(np.float64)
    )
    periods_SMPS_op1.append(
        df_SMPS_op1[i]["P"]["[s]"].to_numpy().astype(np.float64)
    )
    periods_HTRU_low_mid_op1.append(
        df_HTRU_low_mid_op1[i]["P"]["[s]"].to_numpy().astype(np.float64)
    )

    periods_SKA_low_AA4_op1.append(
        df_SKA_low_AA4_op1[i]["P"]["[s]"].to_numpy().astype(np.float64)
    )
    periods_SKA_mid_band1_AA4_op1.append(
        df_SKA_mid_band1_AA4_op1[i]["P"]["[s]"].to_numpy().astype(np.float64)
    )
    periods_SKA_mid_band2_AA4_op1.append(
        df_SKA_mid_band2_AA4_op1[i]["P"]["[s]"].to_numpy().astype(np.float64)
    )

    periods_SKA_low_AAstar_op1.append(
        df_SKA_low_AAstar_op1[i]["P"]["[s]"].to_numpy().astype(np.float64)
    )
    periods_SKA_mid_band1_AAstar_op1.append(
        df_SKA_mid_band1_AAstar_op1[i]["P"]["[s]"]
        .to_numpy()
        .astype(np.float64)
    )
    periods_SKA_mid_band2_AAstar_op1.append(
        df_SKA_mid_band2_AAstar_op1[i]["P"]["[s]"]
        .to_numpy()
        .astype(np.float64)
    )
```

```python
pdots_PMPS_op1 = []
pdots_SMPS_op1 = []
pdots_HTRU_low_mid_op1 = []

pdots_SKA_low_AA4_op1 = []
pdots_SKA_mid_band1_AA4_op1 = []
pdots_SKA_mid_band2_AA4_op1 = []

pdots_SKA_low_AAstar_op1 = []
pdots_SKA_mid_band1_AAstar_op1 = []
pdots_SKA_mid_band2_AAstar_op1 = []

for i in range(run_no):
    pdots_PMPS_op1.append(
        df_PMPS_op1[i]["P_dot"]["[s s^-1]"].to_numpy().astype(np.float64)
    )
    pdots_SMPS_op1.append(
        df_SMPS_op1[i]["P_dot"]["[s s^-1]"].to_numpy().astype(np.float64)
    )
    pdots_HTRU_low_mid_op1.append(
        df_HTRU_low_mid_op1[i]["P_dot"]["[s s^-1]"]
        .to_numpy()
        .astype(np.float64)
    )

    pdots_SKA_low_AA4_op1.append(
        df_SKA_low_AA4_op1[i]["P_dot"]["[s s^-1]"]
        .to_numpy()
        .astype(np.float64)
    )
    pdots_SKA_mid_band1_AA4_op1.append(
        df_SKA_mid_band1_AA4_op1[i]["P_dot"]["[s s^-1]"]
        .to_numpy()
        .astype(np.float64)
    )
    pdots_SKA_mid_band2_AA4_op1.append(
        df_SKA_mid_band2_AA4_op1[i]["P_dot"]["[s s^-1]"]
        .to_numpy()
        .astype(np.float64)
    )

    pdots_SKA_low_AAstar_op1.append(
        df_SKA_low_AAstar_op1[i]["P_dot"]["[s s^-1]"]
        .to_numpy()
        .astype(np.float64)
    )
    pdots_SKA_mid_band1_AAstar_op1.append(
        df_SKA_mid_band1_AAstar_op1[i]["P_dot"]["[s s^-1]"]
        .to_numpy()
        .astype(np.float64)
    )
    pdots_SKA_mid_band2_AAstar_op1.append(
        df_SKA_mid_band2_AAstar_op1[i]["P_dot"]["[s s^-1]"]
        .to_numpy()
        .astype(np.float64)
    )
```

Calculating death lines.

```python
b = 10
```

```python
dl_PMPS_op1 = []
dl_SMPS_op1 = []
dl_HTRU_low_mid_op1 = []

dl_SKA_low_AA4_op1 = []
dl_SKA_mid_band1_AA4_op1 = []
dl_SKA_mid_band2_AA4_op1 = []

dl_SKA_low_AAstar_op1 = []
dl_SKA_mid_band1_AAstar_op1 = []
dl_SKA_mid_band2_AAstar_op1 = []
```

```python
for i in range(run_no):
    dl_PMPS_op1.append(
        2.1e-18 * (1.4 / 1) ** (-1) * b ** (-0.5) * (periods_PMPS_op1[i]) ** 2
    )
    dl_SMPS_op1.append(
        2.1e-18 * (1.4 / 1) ** (-1) * b ** (-0.5) * (periods_SMPS_op1[i]) ** 2
    )
    dl_HTRU_low_mid_op1.append(
        2.1e-18
        * (1.4 / 1) ** (-1)
        * b ** (-0.5)
        * (periods_HTRU_low_mid_op1[i]) ** 2
    )

    dl_SKA_low_AA4_op1.append(
        2.1e-18
        * (1.4 / 1) ** (-1)
        * b ** (-0.5)
        * (periods_SKA_low_AA4_op1[i]) ** 2
    )
    dl_SKA_mid_band1_AA4_op1.append(
        2.1e-18
        * (1.4 / 1) ** (-1)
        * b ** (-0.5)
        * (periods_SKA_mid_band1_AA4_op1[i]) ** 2
    )
    dl_SKA_mid_band2_AA4_op1.append(
        2.1e-18
        * (1.4 / 1) ** (-1)
        * b ** (-0.5)
        * (periods_SKA_mid_band2_AA4_op1[i]) ** 2
    )

    dl_SKA_low_AAstar_op1.append(
        2.1e-18
        * (1.4 / 1) ** (-1)
        * b ** (-0.5)
        * (periods_SKA_low_AAstar_op1[i]) ** 2
    )
    dl_SKA_mid_band1_AAstar_op1.append(
        2.1e-18
        * (1.4 / 1) ** (-1)
        * b ** (-0.5)
        * (periods_SKA_mid_band1_AAstar_op1[i]) ** 2
    )
    dl_SKA_mid_band2_AAstar_op1.append(
        2.1e-18
        * (1.4 / 1) ** (-1)
        * b ** (-0.5)
        * (periods_SKA_mid_band2_AAstar_op1[i]) ** 2
    )
```

### Survey count with a death line

Implementation of death line based on a twisted, multi-polar magnetic field.

Extracting periods and period derivatives:

```python
periods_PMPS_op2 = []
periods_SMPS_op2 = []
periods_HTRU_low_mid_op2 = []

periods_SKA_low_AA4_op2 = []
periods_SKA_mid_band2_AA4_op2 = []

periods_SKA_low_AAstar_op2 = []
periods_SKA_mid_band2_AAstar_op2 = []

for i in range(run_no):
    periods_PMPS_op2.append(
        df_PMPS_op2[i]["P"]["[s]"].to_numpy().astype(np.float64)
    )
    periods_SMPS_op2.append(
        df_SMPS_op2[i]["P"]["[s]"].to_numpy().astype(np.float64)
    )
    periods_HTRU_low_mid_op2.append(
        df_HTRU_low_mid_op2[i]["P"]["[s]"].to_numpy().astype(np.float64)
    )

    periods_SKA_low_AA4_op2.append(
        df_SKA_low_AA4_op2[i]["P"]["[s]"].to_numpy().astype(np.float64)
    )
    periods_SKA_mid_band2_AA4_op2.append(
        df_SKA_mid_band2_AA4_op2[i]["P"]["[s]"].to_numpy().astype(np.float64)
    )

    periods_SKA_low_AAstar_op2.append(
        df_SKA_low_AAstar_op2[i]["P"]["[s]"].to_numpy().astype(np.float64)
    )
    periods_SKA_mid_band2_AAstar_op2.append(
        df_SKA_mid_band2_AAstar_op2[i]["P"]["[s]"]
        .to_numpy()
        .astype(np.float64)
    )
```

```python
pdots_PMPS_op2 = []
pdots_SMPS_op2 = []
pdots_HTRU_low_mid_op2 = []

pdots_SKA_low_AA4_op2 = []
pdots_SKA_mid_band2_AA4_op2 = []

pdots_SKA_low_AAstar_op2 = []
pdots_SKA_mid_band2_AAstar_op2 = []

for i in range(run_no):
    pdots_PMPS_op2.append(
        df_PMPS_op2[i]["P_dot"]["[s s^-1]"].to_numpy().astype(np.float64)
    )
    pdots_SMPS_op2.append(
        df_SMPS_op2[i]["P_dot"]["[s s^-1]"].to_numpy().astype(np.float64)
    )
    pdots_HTRU_low_mid_op2.append(
        df_HTRU_low_mid_op2[i]["P_dot"]["[s s^-1]"]
        .to_numpy()
        .astype(np.float64)
    )

    pdots_SKA_low_AA4_op2.append(
        df_SKA_low_AA4_op2[i]["P_dot"]["[s s^-1]"]
        .to_numpy()
        .astype(np.float64)
    )
    pdots_SKA_mid_band2_AA4_op2.append(
        df_SKA_mid_band2_AA4_op2[i]["P_dot"]["[s s^-1]"]
        .to_numpy()
        .astype(np.float64)
    )

    pdots_SKA_low_AAstar_op2.append(
        df_SKA_low_AAstar_op2[i]["P_dot"]["[s s^-1]"]
        .to_numpy()
        .astype(np.float64)
    )
    pdots_SKA_mid_band2_AAstar_op2.append(
        df_SKA_mid_band2_AAstar_op2[i]["P_dot"]["[s s^-1]"]
        .to_numpy()
        .astype(np.float64)
    )
```

Calculating death lines.

```python
dl_PMPS_op2 = []
dl_SMPS_op2 = []
dl_HTRU_low_mid_op2 = []

dl_SKA_low_AA4_op2 = []
dl_SKA_mid_band2_AA4_op2 = []

dl_SKA_low_AAstar_op2 = []
dl_SKA_mid_band2_AAstar_op2 = []
```

```python
for i in range(run_no):
    dl_PMPS_op2.append(
        2.1e-18 * (1.4 / 1) ** (-1) * b ** (-0.5) * (periods_PMPS_op2[i]) ** 2
    )
    dl_SMPS_op2.append(
        2.1e-18 * (1.4 / 1) ** (-1) * b ** (-0.5) * (periods_SMPS_op2[i]) ** 2
    )
    dl_HTRU_low_mid_op2.append(
        2.1e-18
        * (1.4 / 1) ** (-1)
        * b ** (-0.5)
        * (periods_HTRU_low_mid_op2[i]) ** 2
    )

    dl_SKA_low_AA4_op2.append(
        2.1e-18
        * (1.4 / 1) ** (-1)
        * b ** (-0.5)
        * (periods_SKA_low_AA4_op2[i]) ** 2
    )
    dl_SKA_mid_band2_AA4_op2.append(
        2.1e-18
        * (1.4 / 1) ** (-1)
        * b ** (-0.5)
        * (periods_SKA_mid_band2_AA4_op2[i]) ** 2
    )

    dl_SKA_low_AAstar_op2.append(
        2.1e-18
        * (1.4 / 1) ** (-1)
        * b ** (-0.5)
        * (periods_SKA_low_AAstar_op2[i]) ** 2
    )
    dl_SKA_mid_band2_AAstar_op2.append(
        2.1e-18
        * (1.4 / 1) ** (-1)
        * b ** (-0.5)
        * (periods_SKA_mid_band2_AAstar_op2[i]) ** 2
    )
```

Determining counts of sources with Pdot above the death line Pdot value.

```python
counts_DL_PMPS_op2 = [
    sum(pdot > pdot_dl) for pdot, pdot_dl in zip(pdots_PMPS_op2, dl_PMPS_op2)
]
mean_counts_DL_PMPS_op2 = round_10(np.mean(counts_DL_PMPS_op2))
std_counts_DL_PMPS_op2 = round_10(np.std(counts_DL_PMPS_op2))
print(
    f"PMPS Option 1 with DL: Mean count: {mean_counts_DL_PMPS_op2}, Standard deviation: {std_counts_DL_PMPS_op2}"
)
```

```python
periods_PMPS_op3 = []
periods_SMPS_op3 = []
periods_HTRU_low_mid_op3 = []

periods_SKA_low_AA4_op3 = []
periods_SKA_mid_band2_AA4_op3 = []

periods_SKA_low_AAstar_op3 = []
periods_SKA_mid_band2_AAstar_op3 = []

for i in range(run_no):
    periods_PMPS_op3.append(
        df_PMPS_op3[i]["P"]["[s]"].to_numpy().astype(np.float64)
    )
    periods_SMPS_op3.append(
        df_SMPS_op3[i]["P"]["[s]"].to_numpy().astype(np.float64)
    )
    periods_HTRU_low_mid_op3.append(
        df_HTRU_low_mid_op3[i]["P"]["[s]"].to_numpy().astype(np.float64)
    )

    periods_SKA_low_AA4_op3.append(
        df_SKA_low_AA4_op3[i]["P"]["[s]"].to_numpy().astype(np.float64)
    )
    periods_SKA_mid_band2_AA4_op3.append(
        df_SKA_mid_band2_AA4_op3[i]["P"]["[s]"].to_numpy().astype(np.float64)
    )

    periods_SKA_low_AAstar_op3.append(
        df_SKA_low_AAstar_op3[i]["P"]["[s]"].to_numpy().astype(np.float64)
    )
    periods_SKA_mid_band2_AAstar_op3.append(
        df_SKA_mid_band2_AAstar_op3[i]["P"]["[s]"]
        .to_numpy()
        .astype(np.float64)
    )
```

```python
pdots_PMPS_op3 = []
pdots_SMPS_op3 = []
pdots_HTRU_low_mid_op3 = []

pdots_SKA_low_AA4_op3 = []
pdots_SKA_mid_band2_AA4_op3 = []

pdots_SKA_low_AAstar_op3 = []
pdots_SKA_mid_band2_AAstar_op3 = []

for i in range(run_no):
    pdots_PMPS_op3.append(
        df_PMPS_op3[i]["P_dot"]["[s s^-1]"].to_numpy().astype(np.float64)
    )
    pdots_SMPS_op3.append(
        df_SMPS_op3[i]["P_dot"]["[s s^-1]"].to_numpy().astype(np.float64)
    )
    pdots_HTRU_low_mid_op3.append(
        df_HTRU_low_mid_op3[i]["P_dot"]["[s s^-1]"]
        .to_numpy()
        .astype(np.float64)
    )

    pdots_SKA_low_AA4_op3.append(
        df_SKA_low_AA4_op3[i]["P_dot"]["[s s^-1]"]
        .to_numpy()
        .astype(np.float64)
    )
    pdots_SKA_mid_band2_AA4_op3.append(
        df_SKA_mid_band2_AA4_op3[i]["P_dot"]["[s s^-1]"]
        .to_numpy()
        .astype(np.float64)
    )

    pdots_SKA_low_AAstar_op3.append(
        df_SKA_low_AAstar_op3[i]["P_dot"]["[s s^-1]"]
        .to_numpy()
        .astype(np.float64)
    )
    pdots_SKA_mid_band2_AAstar_op3.append(
        df_SKA_mid_band2_AAstar_op3[i]["P_dot"]["[s s^-1]"]
        .to_numpy()
        .astype(np.float64)
    )
```

Calculating death lines.

```python
dl_PMPS_op3 = []
dl_SMPS_op3 = []
dl_HTRU_low_mid_op3 = []

dl_SKA_low_AA4_op3 = []
dl_SKA_mid_band2_AA4_op3 = []

dl_SKA_low_AAstar_op3 = []
dl_SKA_mid_band2_AAstar_op3 = []
```

```python
for i in range(run_no):
    dl_PMPS_op3.append(
        2.1e-18 * (1.4 / 1) ** (-1) * b ** (-0.5) * (periods_PMPS_op3[i]) ** 2
    )
    dl_SMPS_op3.append(
        2.1e-18 * (1.4 / 1) ** (-1) * b ** (-0.5) * (periods_SMPS_op3[i]) ** 2
    )
    dl_HTRU_low_mid_op3.append(
        2.1e-18
        * (1.4 / 1) ** (-1)
        * b ** (-0.5)
        * (periods_HTRU_low_mid_op3[i]) ** 2
    )

    dl_SKA_low_AA4_op3.append(
        2.1e-18
        * (1.4 / 1) ** (-1)
        * b ** (-0.5)
        * (periods_SKA_low_AA4_op3[i]) ** 2
    )
    dl_SKA_mid_band2_AA4_op3.append(
        2.1e-18
        * (1.4 / 1) ** (-1)
        * b ** (-0.5)
        * (periods_SKA_mid_band2_AA4_op3[i]) ** 2
    )

    dl_SKA_low_AAstar_op3.append(
        2.1e-18
        * (1.4 / 1) ** (-1)
        * b ** (-0.5)
        * (periods_SKA_low_AAstar_op3[i]) ** 2
    )
    dl_SKA_mid_band2_AAstar_op3.append(
        2.1e-18
        * (1.4 / 1) ** (-1)
        * b ** (-0.5)
        * (periods_SKA_mid_band2_AAstar_op3[i]) ** 2
    )
```

Determining counts of sources with Pdot above the death line Pdot value.

```python
counts_DL_PMPS_op3 = [
    sum(pdot > pdot_dl) for pdot, pdot_dl in zip(pdots_PMPS_op3, dl_PMPS_op3)
]
mean_counts_DL_PMPS_op3 = round_10(np.mean(counts_DL_PMPS_op3))
std_counts_DL_PMPS_op3 = round_10(np.std(counts_DL_PMPS_op3))
print(
    f"PMPS Option 1 with DL: Mean count: {mean_counts_DL_PMPS_op3}, Standard deviation: {std_counts_DL_PMPS_op3}"
)
```

```python
counts_DL_SMPS_op3 = [
    sum(pdot > pdot_dl) for pdot, pdot_dl in zip(pdots_SMPS_op3, dl_SMPS_op3)
]
mean_counts_DL_SMPS_op3 = round_10(np.mean(counts_DL_SMPS_op3))
std_counts_DL_SMPS_op3 = round_10(np.std(counts_DL_SMPS_op3))
print(
    f"SMPS Option 1 with DL: Mean count: {mean_counts_DL_SMPS_op3}, Standard deviation: {std_counts_DL_SMPS_op3}"
)
```
