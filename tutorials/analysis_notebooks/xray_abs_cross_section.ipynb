# X-ray absorption cross section
In this notebook we show the energy dependence of the photoelectric absorption cross section for X-ray radiation. 
We compare the cross section computed with variable elemental abundances from [Balucinska-Church and McCammon (1992)](https://ui.adsabs.harvard.edu/abs/1992ApJ...400..699B/abstract) with the approximated cross section with fixed local abundances in the solar neighborood from [Morrison and McCammon (1983)](https://ui.adsabs.harvard.edu/abs/1983ApJ...270..119M/abstract). The logarithmic abundance of an element relative to hydrogen is defined as:
\begin{equation}
A_{\rm el} = \log_{10} \left( \frac{N_{\rm el}}{N_{\rm H}} \right) + 12,
\end{equation}
and by definition the abundance of hydrogen is 12.

```python
import matplotlib.pyplot as plt
import numpy as np
import pypopsyn.simulator.interstellar_medium.xray_abs_cross_section as xabs
import utilities.plot_settings
```

```python
E = np.logspace(np.log10(30), np.log10(10000), 1000)

# Array of abundances relative to hydrogen for 17 elements taken from table 1 in Morrison and McCammon (1983).
abundances_MM83 = np.array(
    [
        12.0,
        11.0,
        8.65,
        7.96,
        8.87,
        8.14,
        6.32,
        7.60,
        6.49,
        7.57,
        7.28,
        5.28,
        6.58,
        6.35,
        5.69,
        7.52,
        6.26,
    ]
)

# Array of abundances relative to hydrogen for 17 elements taken from table 2 in Wilms et al. (2000).
abundances_W00 = np.array(
    [
        12.0,
        10.99,
        8.38,
        7.88,
        8.69,
        7.94,
        6.16,
        7.40,
        6.33,
        7.27,
        7.09,
        5.12,
        6.41,
        6.20,
        5.51,
        7.43,
        6.05,
    ]
)
```

```python
# For better visualization the cross section is multiplied by a factor E^3.
fig, ax = plt.subplots(figsize=(12, 9))

ax.plot(
    E,
    xabs.absorption_cross_section_tot(E, abundances_MM83) * E**3,
    linestyle="-",
    lw=3,
    color="black",
    label="Balucinska-Church et al. (1992) (MM83 abundances)",
)
ax.plot(
    E,
    xabs.absorption_cross_section_tot(E, abundances_W00) * E**3,
    linestyle="-",
    lw=3,
    color="gray",
    label="Balucinska-Church et al. (1992) (W00 abundances)",
)

ax.plot(
    E,
    xabs.absorption_cross_section_approx(E) * E**3,
    linestyle="--",
    lw=3,
    color="tab:red",
    label="Approximated (Morrison and McCammon, 1983)",
)

plt.xlabel(r"$E$ [eV]")
plt.ylabel(r"$\sigma(E) E^3$ [cm$^2$ eV$^2$ / (hydrogen atom)]")
# plt.xlim(0., 50.0)
# plt.ylim(0.1, 2.0e5)
plt.xscale("log")
plt.yscale("log")

plt.legend(frameon=False, loc=0, fontsize=20)

plt.show()
```
